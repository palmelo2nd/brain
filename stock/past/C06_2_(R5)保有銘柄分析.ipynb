### 投資金額別のポートフォリオ

```python
# ============================================================
# 【このセル単体で完結】保有銘柄分析（全体 / 所有者別 / 所有者×証券会社×口座別）
# - merged_df は使わない（このセル内で 02_OWNlist × 01_IDmap × 03_haitou を統合）
# - 集計は「全銘柄」を対象（配当欠損は 0 扱いで年間配当金額を計算）
# - 出力：各ブロックごとに
#   ①投資金額・年間配当・配当利回り（テキスト3行）
#   ②銘柄一覧（現状と同じ項目：code/所有者/証券会社/口座区分/株数/取得単価/name/industry_33/年間配当/年度/区分）
#   ③業種割合（投資金額ベース）円グラフ
# ============================================================

import pandas as pd
import numpy as np
import re
import datetime as _dt
import matplotlib.pyplot as plt
import japanize_matplotlib

# ============================================================
# #(1) 表示設定（省略「...」を出さない）
# ============================================================
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 0)
pd.set_option("display.max_colwidth", None)

# ============================================================
# #(2) 共通：code正規化（文字列のまま）
#   - 前後空白除去
#   - 全角数字→半角
# ============================================================
def normalize_code_str(s: pd.Series) -> pd.Series:
    return (
        s.astype("string")
         .str.strip()
         .str.translate(str.maketrans("０１２３４５６７８９", "0123456789"))
    )

def _to_num_series(x: pd.Series) -> pd.Series:
    # 例: "1,000" や " 100 " を考慮
    s = x.astype("string").str.replace(",", "", regex=False).str.strip()
    return pd.to_numeric(s, errors="coerce")

def _yen(x) -> str:
    if x is None or pd.isna(x):
        return "NaN"
    return f"{int(round(float(x))):,}円"

def _pct(x) -> str:
    if x is None or pd.isna(x):
        return "NaN"
    return f"{float(x)*100:.2f}%"

# ============================================================
# #(3) 02_OWNlist × 01_IDmap を統合（証券会社列も自動判定）
# ============================================================
own_df = pd.read_csv("Data/02_OWNlist.csv", dtype={"code": "string"})
idmap_df = pd.read_csv("Data/01_IDmap.csv", dtype={"code": "string"})

own_df["code"] = normalize_code_str(own_df["code"])
idmap_df["code"] = normalize_code_str(idmap_df["code"])

# --- 証券会社列の自動判定（broker優先、無ければ「証券会社」） ---
if "broker" in own_df.columns:
    broker_col = "broker"
elif "証券会社" in own_df.columns:
    broker_col = "証券会社"
else:
    raise KeyError("02_OWNlist.csv に 'broker' または '証券会社' 列が見つかりません。列名を確認してください。")

# 必須列チェック
need_own = ["code", "所有者", broker_col, "口座区分", "株数", "取得単価"]
need_idm = ["code", "name", "industry_33"]
for c in need_own:
    if c not in own_df.columns:
        raise KeyError(f"02_OWNlist.csv に必要列 '{c}' がありません。")
for c in need_idm:
    if c not in idmap_df.columns:
        raise KeyError(f"01_IDmap.csv に必要列 '{c}' がありません。")

own_df = own_df[need_own].rename(columns={broker_col: "証券会社"})
idmap_df = idmap_df[need_idm]

base_df = pd.merge(own_df, idmap_df, on="code", how="left")

# industry_33 欠損補完（円グラフ用）
base_df["industry_33"] = base_df["industry_33"].fillna("不明")

# ============================================================
# #(4) 03_haitou.csv から「採用する年間配当」を作成
#   ルール：
#   - 予想（"予"付き）があれば、その中で最新年度を採用
#   - 予想が無ければ、実績（"予"なし）の最新年度を採用
#   - 両方無ければ NaN
# ============================================================
haitou_df = pd.read_csv("Data/03_haitou.csv", dtype={"code": "string"}, encoding="utf-8-sig")

# "﻿code" 対策
if "code" not in haitou_df.columns:
    for c in haitou_df.columns:
        if str(c).replace("\ufeff", "") == "code":
            haitou_df = haitou_df.rename(columns={c: "code"})
            break
if "code" not in haitou_df.columns:
    raise KeyError("03_haitou.csv に 'code' 列が見つかりません（BOM等の影響も含め確認してください）。")

haitou_df["code"] = normalize_code_str(haitou_df["code"])

year_cols = [c for c in haitou_df.columns if re.fullmatch(r"\d{4}", str(c))]
year_cols_sorted = sorted(year_cols, key=lambda x: int(x))

def parse_div_cell(x):
    if pd.isna(x):
        return (False, None)
    s = str(x).strip()
    if s == "":
        return (False, None)
    is_forecast = "予" in s
    m = re.search(r"(\d+(?:\.\d+)?)", s)  # 小数対応
    if not m:
        return (is_forecast, None)
    return (is_forecast, float(m.group(1)))

def pick_latest_dividend(row):
    latest_forecast = (None, None)  # (year, value)
    latest_actual = (None, None)
    for y in year_cols_sorted:
        is_fc, val = parse_div_cell(row.get(y))
        if val is None:
            continue
        year_i = int(y)
        if is_fc:
            if latest_forecast[0] is None or year_i > latest_forecast[0]:
                latest_forecast = (year_i, val)
        else:
            if latest_actual[0] is None or year_i > latest_actual[0]:
                latest_actual = (year_i, val)

    if latest_forecast[0] is not None:
        return pd.Series({"年間配当": latest_forecast[1], "年間配当_年度": latest_forecast[0], "年間配当_区分": "予想"})
    if latest_actual[0] is not None:
        return pd.Series({"年間配当": latest_actual[1], "年間配当_年度": latest_actual[0], "年間配当_区分": "実績"})
    return pd.Series({"年間配当": pd.NA, "年間配当_年度": pd.NA, "年間配当_区分": pd.NA})

picked = haitou_df.apply(pick_latest_dividend, axis=1)
haitou_pick_df = pd.concat([haitou_df[["code"]].copy(), picked], axis=1)

# codeごとに1行に整理（同一codeが複数行ある想定の保険）
haitou_pick_df = (
    haitou_pick_df
    .sort_values(["code", "年間配当_年度"])
    .drop_duplicates(subset=["code"], keep="last")
)

# ============================================================
# #(5) base_df に配当列を追加 → analyze_df
# ============================================================
analyze_df = pd.merge(base_df, haitou_pick_df, on="code", how="left")

# ============================================================
# #(6) 投資金額・年間配当金額・利回り列を作成（全銘柄対象）
#   + Data/jyouto/summary.csv の「総合実現損益（2行目）」を紐付け
#   + 配当利回りの分母は「所有者=T」のときだけ実現損益反映（投資金額−実現損益）
# ============================================================
analyze_df["株数_num"] = _to_num_series(analyze_df["株数"])
analyze_df["取得単価_num"] = _to_num_series(analyze_df["取得単価"])
analyze_df["年間配当_num"] = pd.to_numeric(analyze_df["年間配当"], errors="coerce")

analyze_df["投資金額"] = analyze_df["株数_num"] * analyze_df["取得単価_num"]
analyze_df["年間配当金額"] = analyze_df["株数_num"] * analyze_df["年間配当_num"].fillna(0)

# --- 実現損益（総合）を summary.csv から取り込み（2行目） ---
def _to_num_scalar(x) -> float:
    """'1,234' や '12,345円' 等を数値化（失敗時は NaN）"""
    if pd.isna(x):
        return float("nan")
    s = str(x).replace(",", "").strip()
    m = re.search(r"-?\d+(?:\.\d+)?", s)
    if not m:
        return float("nan")
    try:
        return float(m.group(0))
    except Exception:
        return float("nan")

summary_path = "Data/jyouto/summary.csv"
summary_df = pd.read_csv(summary_path, dtype="string", encoding="utf-8-sig")
if len(summary_df) < 2:
    raise ValueError(f"{summary_path} は2行以上必要です（2行目に総合実現損益が入る想定）。")

row2 = summary_df.iloc[1]  # 2行目（index=1）

pl_map_by_code: dict[str, float] = {}
pl_map_by_name: dict[str, float] = {}

for k, v in row2.items():
    if k is None:
        continue
    key_raw = str(k).strip()
    if key_raw == "":
        continue

    val_num = _to_num_scalar(v)
    if pd.isna(val_num):
        continue

    key_norm = normalize_code_str(pd.Series([key_raw], dtype="string")).iloc[0]
    if re.fullmatch(r"\d{4}", str(key_norm)):
        pl_map_by_code[str(key_norm)] = float(val_num)
    else:
        pl_map_by_name[key_raw] = float(val_num)

analyze_df["code_norm"] = normalize_code_str(analyze_df["code"].astype("string"))
analyze_df["総合実現損益"] = analyze_df["code_norm"].map(pl_map_by_code)

# codeで取れない場合は name でも補完（name列がある場合のみ）
if "name" in analyze_df.columns and len(pl_map_by_name) > 0:
    analyze_df["総合実現損益"] = analyze_df["総合実現損益"].fillna(analyze_df["name"].map(pl_map_by_name))

# 未該当は0扱い
analyze_df["総合実現損益"] = pd.to_numeric(analyze_df["総合実現損益"], errors="coerce").fillna(0.0)

# --- 利回り計算用の投資金額（所有者=Tのみ実現損益反映） ---
analyze_df["投資金額_利回り用"] = np.where(
    analyze_df["所有者"].astype("string") == "T",
    analyze_df["投資金額"] - analyze_df["総合実現損益"],
    analyze_df["投資金額"],
)

# 配当利回り（分母は投資金額_利回り用）
analyze_df["配当利回り"] = np.where(
    (analyze_df["投資金額_利回り用"].notna()) & (analyze_df["投資金額_利回り用"] > 0),
    analyze_df["年間配当金額"] / analyze_df["投資金額_利回り用"],
    np.nan
)

# ============================================================
# #(7) 表示：テキスト2行（投資金額・年間配当） → 一覧表 → 円グラフ
#   - 配当利回りは「計算しない＆表示しない」
#   - 一覧表の列：code, name, 所有者, 口座区分, 株数, 取得単価, 配当, 配当年度, 業種, 実現損益
#     ※ 実現損益は所有者=Tのときのみ。それ以外は 0 を表示
#   - 円グラフは「投資金額が計算できる行のみ」で投資金額ベース業種割合
# ============================================================
DISPLAY_COLS_RENAMED = [
    "code", "name", "所有者", "口座区分", "株数", "取得単価",
    "配当", "配当年度", "業種", "実現損益"
]

def show_portfolio_block(df: pd.DataFrame, title: str):
    print("\n" + "=" * 100)
    print(title)
    print("=" * 100)

    # 集計・円グラフ用：投資金額が計算できるものだけ
    df_valid_inv = df[df["投資金額"].notna() & (df["投資金額"] > 0)].copy()

    total_inv = df_valid_inv["投資金額"].sum()
    total_div = df_valid_inv["年間配当金額"].sum()

    # ★ テキストは 2 行のみ（利回りは出さない）
    print(f"投資金額: {_yen(total_inv)}")
    print(f"年間配当: {_yen(total_div)}")

    # --------------------------------------------------------
    # 一覧表（列名・順番・実現損益表示ルールを反映）
    # --------------------------------------------------------
    disp = df.copy()

    # 表示用：列名を揃える
    if "年間配当" in disp.columns:
        disp["配当"] = disp["年間配当"]
    else:
        disp["配当"] = pd.NA

    if "年間配当_年度" in disp.columns:
        disp["配当年度"] = disp["年間配当_年度"]
    else:
        disp["配当年度"] = pd.NA

    if "industry_33" in disp.columns:
        disp["業種"] = disp["industry_33"]
    else:
        disp["業種"] = pd.NA

    # 実現損益：所有者=T のときだけ。その他は 0
    if "総合実現損益" in disp.columns:
        disp["実現損益"] = np.where(
            disp["所有者"].astype("string") == "T",
            pd.to_numeric(disp["総合実現損益"], errors="coerce").fillna(0.0),
            0.0
        )
    else:
        disp["実現損益"] = 0.0

    # 必要列だけ・順序通り（欠損列があっても落ちない保険）
    cols = [c for c in DISPLAY_COLS_RENAMED if c in disp.columns]
    display(disp[cols])

    # --------------------------------------------------------
    # 円グラフ（投資金額ベース業種割合）
    # --------------------------------------------------------
    pie_src = (
        df_valid_inv.groupby("industry_33", dropna=False)["投資金額"]
        .sum()
        .sort_values(ascending=False)  # ★ 降順
    )

    if len(pie_src) == 0:
        print("円グラフ: 投資金額が計算できる銘柄がありません（株数/取得単価の欠損・形式を確認してください）。")
        return

    plt.figure()
    plt.title(f"{title}\n（投資金額ベース業種割合 / 合計: {_yen(total_inv)}）")
    plt.pie(
        pie_src.values,
        labels=pie_src.index,
        autopct="%1.1f%%",
        startangle=90,        # ★ 12時（0時）方向から開始
        counterclock=False    # ★ 時計回り
    )
    plt.axis("equal")
    plt.show()

# ============================================================
# #(8) (1)全体 → (2)所有者別
# ============================================================
show_portfolio_block(analyze_df, "【(1) 全体】")

for owner, df_owner in analyze_df.groupby("所有者", dropna=False):
    show_portfolio_block(df_owner, f"【(2) 所有者別】所有者={owner}")
```

### 配当銘柄分析

```python
# ============================================================
# 【このセル単体で完結】保有銘柄 → 条件抽出(target_df) → 配当分析（円グラフ2列＋横棒＋スコア）
#
# 入力CSV:
#   Data/02_OWNlist.csv  : code, 所有者, (broker or 証券会社), 口座区分, 株数, 取得単価
#   Data/01_IDmap.csv    : code, name, industry_33
#   Data/03_haitou.csv   : code, 2019, 2020, ... （セルに「予」あり/なし）
#
# 抽出条件（targets）:
#   - 年間配当があり、かつ 年間配当_年度 が「今年 or 昨年」
#   - (所有者, 証券会社, 口座区分) の組が targets に一致（None または "*" はワイルドカード）
#
# 出力:
#   - 全体：円グラフ（投資金額/配当金額）＋横棒（配当金額ベース%）＋スコア（円グラフ右に表示）
#   - 所有者別：同様
# ============================================================

import pandas as pd
import numpy as np
import re
import datetime as _dt
import matplotlib.pyplot as plt
import japanize_matplotlib

# ============================================================
# #(1) 表示設定（省略「...」を出さない）
# ============================================================
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 0)
pd.set_option("display.max_colwidth", None)

# ============================================================
# #(2) 共通ユーティリティ
# ============================================================
def normalize_code_str(s: pd.Series) -> pd.Series:
    return (
        s.astype("string")
         .str.strip()
         .str.translate(str.maketrans("０１２３４５６７８９", "0123456789"))
    )

def _to_num_series(x: pd.Series) -> pd.Series:
    s = x.astype("string").str.replace(",", "", regex=False).str.strip()
    return pd.to_numeric(s, errors="coerce")

def _to_num_scalar(x) -> float:
    """'1,234' や ' 123 ' のような表記を数値化（失敗時は NaN）"""
    if pd.isna(x):
        return float("nan")
    s = str(x).replace(",", "").strip()
    # "円" 等が混ざっていても数値だけ拾う
    m = re.search(r"-?\d+(?:\.\d+)?", s)
    if not m:
        return float("nan")
    try:
        return float(m.group(0))
    except Exception:
        return float("nan")

def _yen(x: float) -> str:
    if pd.isna(x):
        return "0円"
    return f"{int(round(float(x))):,}円"

def _pct_series_by_industry(work_df: pd.DataFrame, value_col: str) -> pd.Series:
    s = work_df.groupby("industry_33")[value_col].sum()
    total = float(s.sum())
    if total <= 0:
        return pd.Series(dtype=float)
    return (s / total) * 100.0

# ============================================================
# #(3) CSV読み込み & 02_OWNlist × 01_IDmap 統合（merged_df相当をセル内で作る）
#   ★変更点：
#     - 01_IDmap.csv から L_def_score も（存在すれば）取り込む
#     - L_def_score が無い場合は merged_df["L_def_score"] を NA で作る
#     - L_def（手作業ラベル）は読んでもよいが、今回のdefensive判定には使わない
# ============================================================
own_df = pd.read_csv("Data/02_OWNlist.csv", dtype={"code": "string"})
idmap_df = pd.read_csv("Data/01_IDmap.csv", dtype={"code": "string"}, encoding="utf-8-sig")

own_df["code"] = normalize_code_str(own_df["code"])
idmap_df["code"] = normalize_code_str(idmap_df["code"])

# --- 証券会社列の自動判定（broker優先、無ければ「証券会社」） ---
if "broker" in own_df.columns:
    broker_col = "broker"
elif "証券会社" in own_df.columns:
    broker_col = "証券会社"
else:
    raise KeyError("02_OWNlist.csv に 'broker' または '証券会社' 列が見つかりません。列名を確認してください。")

need_own = ["code", "所有者", broker_col, "口座区分", "株数", "取得単価"]

# ★ L_def / L_def_score は任意（あれば取り込む）
need_idm_base = ["code", "name", "industry_33"]
need_idm = need_idm_base.copy()
if "L_def" in idmap_df.columns:
    need_idm.append("L_def")
if "L_def_score" in idmap_df.columns:
    need_idm.append("L_def_score")

for c in need_own:
    if c not in own_df.columns:
        raise KeyError(f"02_OWNlist.csv に必要列 '{c}' がありません。")
for c in need_idm_base:
    if c not in idmap_df.columns:
        raise KeyError(f"01_IDmap.csv に必要列 '{c}' がありません。")

own_df = own_df[need_own].rename(columns={broker_col: "証券会社"})
idmap_df = idmap_df[need_idm]

merged_df = pd.merge(own_df, idmap_df, on="code", how="left")

# L_def_score が無い場合に備えて列だけ作る（以降のスコアで参照するため）
if "L_def_score" not in merged_df.columns:
    merged_df["L_def_score"] = pd.NA

# 参考：L_def が無い場合でも列は作っておく（ただし今回のdefensive判定はこれを使わない）
if "L_def" not in merged_df.columns:
    merged_df["L_def"] = pd.NA

# ============================================================
# #(4) 03_haitou.csv から「採用する年間配当」を作成して merged_df に付与
# ============================================================
haitou_df = pd.read_csv("Data/03_haitou.csv", dtype={"code": "string"}, encoding="utf-8-sig")

# "﻿code" 対策
if "code" not in haitou_df.columns:
    for c in haitou_df.columns:
        if str(c).replace("\ufeff", "") == "code":
            haitou_df = haitou_df.rename(columns={c: "code"})
            break
if "code" not in haitou_df.columns:
    raise KeyError("03_haitou.csv に 'code' 列が見つかりません（BOM等を確認）。")

haitou_df["code"] = normalize_code_str(haitou_df["code"])

year_cols = [c for c in haitou_df.columns if re.fullmatch(r"\d{4}", str(c))]
year_cols_sorted = sorted(year_cols, key=lambda x: int(x))

def _parse_div_cell(x):
    if pd.isna(x):
        return (False, None)
    s = str(x).strip()
    if s == "":
        return (False, None)
    is_forecast = "予" in s
    m = re.search(r"(\d+(?:\.\d+)?)", s)  # 小数対応
    if not m:
        return (is_forecast, None)
    return (is_forecast, float(m.group(1)))

def _pick_latest_dividend(row):
    latest_forecast = (None, None)  # (year, value)
    latest_actual = (None, None)

    for y in year_cols_sorted:
        is_fc, val = _parse_div_cell(row.get(y))
        if val is None:
            continue
        yi = int(y)
        if is_fc:
            if latest_forecast[0] is None or yi > latest_forecast[0]:
                latest_forecast = (yi, val)
        else:
            if latest_actual[0] is None or yi > latest_actual[0]:
                latest_actual = (yi, val)

    if latest_forecast[0] is not None:
        return pd.Series({"年間配当": latest_forecast[1], "年間配当_年度": latest_forecast[0], "年間配当_区分": "予想"})
    if latest_actual[0] is not None:
        return pd.Series({"年間配当": latest_actual[1], "年間配当_年度": latest_actual[0], "年間配当_区分": "実績"})
    return pd.Series({"年間配当": pd.NA, "年間配当_年度": pd.NA, "年間配当_区分": pd.NA})

picked = haitou_df.apply(_pick_latest_dividend, axis=1)
haitou_pick_df = pd.concat([haitou_df[["code"]].copy(), picked], axis=1)
haitou_pick_df = (
    haitou_pick_df
    .sort_values(["code", "年間配当_年度"])
    .drop_duplicates(subset=["code"], keep="last")
)

merged_df = pd.merge(merged_df, haitou_pick_df, on="code", how="left")

# ============================================================
# #(5) 抽出関数（ユーザー提示のロジック）
# ============================================================
def filter_target_stocks(
    merged_df: pd.DataFrame,
    owners_brokers_accounts: list[tuple],
) -> pd.DataFrame:

    CURRENT_YEAR = _dt.date.today().year
    TARGET_YEARS = {CURRENT_YEAR, CURRENT_YEAR - 1}

    base_df = merged_df[
        merged_df["年間配当"].notna()
        & merged_df["年間配当_年度"].notna()
        & merged_df["年間配当_年度"].astype("Int64").isin(TARGET_YEARS)
    ].copy()

    cond = False
    for owner, broker, account in owners_brokers_accounts:
        c = True
        if owner not in (None, "*"):
            c = c & (base_df["所有者"] == owner)
        if broker not in (None, "*"):
            c = c & (base_df["証券会社"] == broker)
        if account not in (None, "*"):
            c = c & (base_df["口座区分"] == account)
        cond = cond | c

    return base_df[cond].copy()

# 対象口座の指定（必要に応じて編集）
targets = [
    ("Y", "SBI", "NISA"),
    ("Y", "SBI", "特定"),
    ("T", "SBI", "NISA"),
    ("T", "SBI", "特定"),
]

target_df = filter_target_stocks(merged_df, targets)

print("=== 条件一致・集計対象銘柄 ===")
print("件数:", len(target_df))
display(target_df)

# ============================================================
# #(6) 事前整形：金額列作成（ユーザー元コードに合わせる）
#   + Data/jyouto/summary.csv の「総合実現損益（2行目）」を紐付け
#   + 実現損益は所有者=Tのときのみ扱う（表示も計算も）
#   + (補*)用：
#       - T かつ 実現損益>0 のときだけ投資金額を減算して利回り計算に反映
#       - 実現損益<=0（損失）は利回り計算に反映しない
#   + 表出力用の列：
#       code, name, 所有者, 口座区分, 株数, 取得単価, 取得単価(補), 配当, 年間配当(補),
#       配当年度, 業種, 実現損益
# ============================================================
df = target_df.copy()

df["株数"] = _to_num_series(df["株数"])
df["取得単価"] = _to_num_series(df["取得単価"])
df["年間配当"] = pd.to_numeric(df["年間配当"], errors="coerce")

df["投資金額"] = df["株数"] * df["取得単価"]
df["配当金額"] = df["株数"] * df["年間配当"]

# --- 実現損益（総合）を summary.csv から取り込み ---
summary_path = "Data/jyouto/summary.csv"
summary_df = pd.read_csv(summary_path, dtype="string", encoding="utf-8-sig")

if len(summary_df) < 2:
    raise ValueError(f"{summary_path} は2行以上必要です（2行目に総合実現損益が入る想定）。")

row2 = summary_df.iloc[1]  # 2行目（index=1）

pl_map_by_code: dict[str, float] = {}
pl_map_by_name: dict[str, float] = {}

for k, v in row2.items():
    if k is None:
        continue
    key_raw = str(k).strip()
    if key_raw == "":
        continue

    val_num = _to_num_scalar(v)
    if pd.isna(val_num):
        continue

    key_norm = (
        pd.Series([key_raw], dtype="string")
          .pipe(normalize_code_str)
          .iloc[0]
    )
    if re.fullmatch(r"\d{4}", str(key_norm)):
        pl_map_by_code[str(key_norm)] = float(val_num)
    else:
        pl_map_by_name[key_raw] = float(val_num)

# df へ紐付け（code優先、なければ name で補完）
df["code_norm"] = normalize_code_str(df["code"])
df["総合実現損益"] = df["code_norm"].map(pl_map_by_code)

if "name" in df.columns and len(pl_map_by_name) > 0:
    df["総合実現損益"] = df["総合実現損益"].fillna(df["name"].map(pl_map_by_name))

df["総合実現損益"] = pd.to_numeric(df["総合実現損益"], errors="coerce").fillna(0.0)

# --- 実現損益は所有者=Tのときのみ扱う（表示用） ---
df["実現損益"] = np.where(
    df["所有者"].astype("string") == "T",
    df["総合実現損益"],
    0.0
)

# --- (補*)：利回り計算に反映する実現益（T かつ >0 のときのみ） ---
df["実現益_利回り反映"] = np.where(
    (df["所有者"].astype("string") == "T") & (df["総合実現損益"] > 0),
    df["総合実現損益"],
    0.0
)

# --- (補*) 投資金額（分母） ---
# 例：10万円投資・実現益2万円 → 8万円を投資しているものとして利回り計算
# 損失（負）は反映しない
df["投資金額_補"] = df["投資金額"] - df["実現益_利回り反映"]

# --- 表用：取得単価(補) = 投資金額_補 / 株数 ---
df["取得単価(補)"] = np.where(
    (df["株数"].notna()) & (df["株数"] > 0) & (df["投資金額_補"].notna()),
    df["投資金額_補"] / df["株数"],
    np.nan
)

# --- 表用：配当（表示名） ---
df["配当"] = df["年間配当"]

# --- 表用：年間配当(補)
# 実現益で取得原価（投資金額_補）が減ると、同じ配当でも効きが上がる
# 年間配当(補) = 年間配当 * (投資金額 / 投資金額_補)（投資金額_補>0のとき）
df["年間配当(補)"] = np.where(
    (df["投資金額_補"].notna()) & (df["投資金額_補"] > 0) & (df["投資金額"].notna()) & (df["投資金額"] > 0),
    df["年間配当"] * (df["投資金額"] / df["投資金額_補"]),
    df["年間配当"]
)

# --- (補*) 棒グラフ用：配当金額_補 ---
df["配当金額_補"] = df["株数"] * df["年間配当(補)"]

# --- 業種が無い/ダミーは除外（円グラフの分母を汚さない） ---
valid_industry = (
    df["industry_33"].notna()
    & ~df["industry_33"].astype("string").str.strip().isin(["", "-", "0"])
)
df = df[valid_industry].copy()

# --- 表出力（列順・文言を指定通りに） ---
df["配当年度"] = df["年間配当_年度"]
df["業種"] = df["industry_33"]

DISPLAY_COLS = [
    "code", "name", "所有者", "口座区分", "株数", "取得単価", "取得単価(補)",
    "配当", "年間配当(補)", "配当年度", "業種", "実現損益"
]
print("=== 条件一致・集計対象銘柄（表示用） ===")
print("件数:", len(df))
display(df[DISPLAY_COLS])

# ============================================================
# #(7) 棒グラフの軸：01_IDmap.csv の industry_33 全種類（表記ゆれ防止）
#   ★変更点：
#     - 「日経ETF」「米国ETF」も評価対象に含める（ただし15%超でも減点しない対象）
# ============================================================
if "industry_33" not in idmap_df.columns:
    raise KeyError("01_IDmap.csv に 'industry_33' 列が見つかりません。")

all_categories = (
    idmap_df["industry_33"]
    .astype("string")
    .map(lambda v: v.strip() if isinstance(v, str) else v)
    .dropna()
)
all_categories = sorted({v for v in all_categories.unique().tolist() if v not in ["", "-", "0"]})

# ★ 以前の「日経ETF」「米国ETF」除外は撤廃（スコア要件変更のため）
EXCLUDED_INDUSTRIES = set()

all_categories_eval = [c for c in all_categories if c not in EXCLUDED_INDUSTRIES]
if len(all_categories_eval) == 0:
    raise ValueError("評価対象の industry_33 が 0 件になりました。除外設定やIDmapを確認してください。")

# ============================================================
# #(8) スコア計算（利回り/銘柄集中は「投資金額_補」を使用）
#   ★変更点（今回の要件）：
#     ④ defensiveスコア（40点満点）は「元の方式」に戻す
#        - L_def_score >= 60 を L_def=1 とみなす（擬似ラベル）
#        - その L_def=1 の配当金額比率(%) を defensive比率とする
#        - 0%→0点、50%→40点、50%以上は40点キープ（線形補完）
# ============================================================
OTHER_PRODUCTS = "その他製品"
NON_PENALTY_INDUSTRIES = {OTHER_PRODUCTS, "J-REIT", "日経ETF", "米国ETF"}

DEF_SCORE_TH = 60.0  # ★def_score(=L_def_score) の閾値：>=60 を defensive とみなす

def _score_dividend_yield(total_invest: float, total_dividend: float) -> tuple[float, float]:
    """
    利回り(%)とスコア(0-100)を返す
    4%→0点、10%→100点の線形（範囲外はクリップ）
    """
    if total_invest <= 0:
        return 0.0, 0.0
    y = (total_dividend / total_invest) * 100.0

    y0, y1 = 4.0, 10.0
    score = (y - y0) / (y1 - y0) * 100.0
    score = max(0.0, min(100.0, float(score)))
    return float(y), float(score)


def _score_industries_by_dividend_share(work_df: pd.DataFrame) -> tuple[float, list[str]]:
    """
    業種スコア(0-40)とメッセージ配列を返す
    - 配当金額（補正なし: 配当金額）ベース
    - 0〜1.5%：0点→満点へ線形
      1.5〜15%：満点
      15〜17%：線形減点
      17%以上：0点
    - ただし NON_PENALTY_INDUSTRIES（その他製品 / J-REIT / 日経ETF / 米国ETF）は
      15%超でも減点しない（満点キープ）
    - 注意メッセージは不要（常に空配列）
    """
    msgs: list[str] = []  # ★ 常に空（要望）

    total_dividend = float(work_df["配当金額"].sum())
    if total_dividend <= 0:
        return 0.0, msgs

    share = work_df.groupby("industry_33")["配当金額"].sum() / total_dividend * 100.0
    share = share.reindex(all_categories_eval, fill_value=0.0)

    n = len(all_categories_eval)
    per_max = 40.0 / n  # 40点を業種数で均等配分

    def piecewise_points_general(r_pct: float) -> float:
        r = float(r_pct)
        if r <= 0.0:
            return 0.0
        if 0.0 < r < 1.5:
            return per_max * (r / 1.5)
        if 1.5 <= r <= 15.0:
            return per_max
        if 15.0 < r < 17.0:
            return per_max * (1.0 - (r - 15.0) / 2.0)
        return 0.0

    def piecewise_points_no_penalty_over15(r_pct: float) -> float:
        r = float(r_pct)
        if r <= 0.0:
            return 0.0
        if 0.0 < r < 1.5:
            return per_max * (r / 1.5)
        return per_max  # 1.5%以上は満点キープ

    score = 0.0
    for ind, r in share.items():
        if ind in NON_PENALTY_INDUSTRIES:
            score += piecewise_points_no_penalty_over15(r)
        else:
            score += piecewise_points_general(r)

    score = max(0.0, min(40.0, float(score)))
    return score, msgs
    

def _score_stocks_by_invest_concentration(work_df: pd.DataFrame) -> float:
    total_invest = float(work_df["投資金額_補"].sum())
    if total_invest <= 0:
        return 0.0

    by_code = work_df.groupby("code")["投資金額_補"].sum()
    weights = by_code / total_invest

    excess = (weights - 0.03).clip(lower=0.0)
    penalty_points = float((excess * 100.0).sum())

    score = 20.0 - penalty_points
    return max(0.0, float(score))


def _score_defensive_by_dividend_share_from_defscore(work_df: pd.DataFrame) -> tuple[float, float]:
    """
    defensive比率(%)とスコア(0-40)を返す（元方式）

    - 配当金額（補正なし: 配当金額）ベース
    - L_def_score >= DEF_SCORE_TH を L_def=1（defensive）とみなす
    - 0%→0点、50%→40点、50%以上は40点キープ（線形）
    """
    total_dividend = float(work_df["配当金額"].sum())
    if total_dividend <= 0:
        return 0.0, 0.0

    s = pd.to_numeric(work_df.get("L_def_score", pd.Series([pd.NA]*len(work_df))), errors="coerce")
    defensive_mask = s >= float(DEF_SCORE_TH)

    def_div = float(work_df.loc[defensive_mask, "配当金額"].sum())
    ratio = (def_div / total_dividend) * 100.0

    score = (ratio / 50.0) * 40.0
    score = max(0.0, min(40.0, float(score)))
    return float(ratio), float(score)


def _calc_portfolio_score(work_df: pd.DataFrame) -> dict:
    total_invest = float(work_df["投資金額"].sum())
    total_invest_sup = float(work_df["投資金額_補"].sum())
    total_dividend = float(work_df["配当金額"].sum())

    y_pct, _ = _score_dividend_yield(total_invest, total_dividend)
    y_sup_pct, score_y = _score_dividend_yield(total_invest_sup, total_dividend)

    score_i, ind_msgs = _score_industries_by_dividend_share(work_df)
    score_s = _score_stocks_by_invest_concentration(work_df)

    def_ratio_pct, score_d = _score_defensive_by_dividend_share_from_defscore(work_df)

    score_total = score_y + score_i + score_s + score_d
    score_max = 100.0 + 40.0 + 20.0 + 40.0  # 200点満点

    return {
        "total_invest": total_invest,
        "total_invest_sup": total_invest_sup,
        "total_dividend": total_dividend,
        "dividend_yield_pct": y_pct,
        "dividend_yield_sup_pct": y_sup_pct,  # (補*)
        "score_yield": score_y,               # /100
        "score_industry": score_i,            # /40
        "score_stock": score_s,               # /20

        # ★ defensiveは元方式に戻す
        "defensive_ratio_pct": def_ratio_pct, # %
        "score_defensive": score_d,           # /40

        "score_total": score_total,
        "score_max": score_max,
        "messages": ind_msgs,
    }

# ============================================================
# #(9) 円グラフ：投資金額ベース & 配当金額ベース（右にスコア内訳表示）
#   ★変更点：
#     - defensive比率は「L_def_score>=60 を defensive とみなした配当比率」を表示
# ============================================================
def _plot_two_pies(work_df: pd.DataFrame, title_prefix: str):
    fig, axes = plt.subplots(1, 2, figsize=(14, 7))

    info = _calc_portfolio_score(work_df)

    for ax, value_col, subtitle in zip(
        axes,
        ["投資金額_補", "配当金額"],
        ["投資金額ベース（補*）", "配当金額ベース"],
    ):
        by_ind = work_df.groupby("industry_33")[value_col].sum()
        by_ind = by_ind[by_ind > 0].sort_values(ascending=False)

        total = float(by_ind.sum())

        if len(by_ind) == 0:
            ax.text(0.5, 0.5, "データなし", ha="center", va="center")
            ax.set_title(f"{subtitle}\n（合計 {_yen(0)}）")
            ax.axis("off")
            continue

        ax.pie(
            by_ind.values,
            labels=by_ind.index,
            autopct="%1.1f%%",
            startangle=90,
            counterclock=False,
        )
        ax.axis("equal")

        if value_col == "配当金額":
            ax.set_title(
                f"{subtitle}\n"
                f"（合計 {_yen(total)} / 配当利回り {info['dividend_yield_pct']:.2f}%"
                f" / 配当利回り(補*) {info['dividend_yield_sup_pct']:.2f}%"
                f" / defensive比率 {info['defensive_ratio_pct']:.1f}%）"
            )

            lines = [
                f"スコア {info['score_total']:.1f} / {info['score_max']:.0f}",
                f"配当利回り {info['score_yield']:.1f} / 100",
                f"業種 {info['score_industry']:.1f} / 40",
                f"銘柄 {info['score_stock']:.1f} / 20",
                f"defensive {info['score_defensive']:.1f} / 40",
            ]
            for m in info.get("messages", []):
                lines.append(m)

            ax.text(0, 0, "\n".join(lines), ha="center", va="center", fontsize=10)
        else:
            ax.set_title(f"{subtitle}\n（合計 {_yen(total)}）")

    fig.suptitle(title_prefix, fontsize=14)
    plt.tight_layout()
    plt.show()

# ============================================================
# #(10) 横棒グラフ：業種割合（%）
#   - (補*) の計算結果を反映：配当金額_補 ベースで業種割合を算出
#   - 01_IDmap.csv の industry_33 全種類を軸に（該当なしは0%）
#   - 「日経ETF」「米国ETF」は除外（棒グラフ/業種スコアから外す）
# ============================================================
def _plot_dividend_barh_pct(work_df: pd.DataFrame, title_prefix: str):
    pct = _pct_series_by_industry(work_df, "配当金額_補")
    pct = pct.reindex(all_categories_eval, fill_value=0.0)

    # barhは昇順にすると「下が最大」
    pct = pct.sort_values(ascending=True)

    fig_height = max(8, len(pct) * 0.35)
    plt.figure(figsize=(9, fig_height))
    plt.barh(pct.index, pct.values)
    plt.title(f"{title_prefix}：配当金額ベース 業種割合（補* / %）")
    plt.xlabel("割合（%）")
    plt.xlim(0, max(1.0, float(pct.max()) * 1.1))
    plt.tight_layout()
    plt.show()

# ============================================================
# #(11) 実行：全体 → 所有者別
# ============================================================
_plot_two_pies(df, "全体：業種割合（円グラフ）")
_plot_dividend_barh_pct(df, "全体")

owners = sorted(df["所有者"].astype("string").dropna().unique().tolist())
for owner in owners:
    dfo = df[df["所有者"] == owner].copy()
    _plot_two_pies(dfo, f"所有者 {owner}：業種割合（円グラフ）")
    _plot_dividend_barh_pct(dfo, f"所有者 {owner}")
```

### 推奨銘柄

```python
# ============================================================
# 【このセル単体で完結】推奨銘柄提案（関数化／所有者指定対応／日本語列／スコア内訳列）
#
# 使い方：
#   propose_recommendations()            # 全体（target_df全行）に対して提案
#   propose_recommendations("Y")         # 所有者Yの保有分に対して提案
#   propose_recommendations("T", topn=50)
#
# 前提：
#   - 直前セルで target_df が作成済み
#   - target_df 必要列：code, industry_33, 株数, 取得単価, 年間配当, 所有者
#
# 候補銘柄：
#   - Data/01_IDmap.csv の L_高配当 == 1
#
# 株価：
#   - Data/kabuka/YYYY.csv（Data/kabuka内の最大年CSV）の各銘柄列から「最後の非NaN終値」を採用
#
# 追加購入ルール：
#   - 100株単位
#   - 100株の投資金額が20万円以下なら、20万円を超える最小の100株単位まで増やす
#
# スコア：
#   -（貼り直し版の2つ目コードと同仕様）
#     配当利回り 40点（4%→0点、10%→40点）
#     業種 40点（配当金額比率 piecewise / ETF2業種除外 / 40÷業種数）
#     銘柄 20点（投資金額比率 3%超過分[%]を減点）
# ============================================================

# ============================================================
# #(1) imports
# ============================================================
import os
import re
import math
import pandas as pd
import numpy as np


# ============================================================
# #(2) 共通ユーティリティ
# ============================================================
EXCLUDED_INDUSTRIES = {"日経ETF", "米国ETF"}

# ★ defensive判定は def_score（=L_def_score）>=60 を L_def=1 とみなす
DEF_SCORE_TH = 60.0

def normalize_code_str_series(s: pd.Series) -> pd.Series:
    return (
        s.astype("string")
         .str.strip()
         .str.translate(str.maketrans("０１２３４５６７８９", "0123456789"))
    )

def _to_num_scalar(x) -> float:
    """'1,234' や ' 123 ' のような表記を数値化（失敗時は NaN）"""
    if pd.isna(x):
        return float("nan")
    s = str(x).replace(",", "").strip()
    m = re.search(r"-?\d+(?:\.\d+)?", s)
    if not m:
        return float("nan")
    try:
        return float(m.group(0))
    except Exception:
        return float("nan")

def _ldef_from_defscore_series(defscore_s: pd.Series, th: float = DEF_SCORE_TH) -> pd.Series:
    """
    L_def_score（def_score）から自動L_def(0/1)を作る
      - def_score >= th -> 1
      - def_score <  th -> 0
      - NaN は 0（= offensive）扱い（必要なら 0.5 など中立に変える）
    """
    x = pd.to_numeric(defscore_s, errors="coerce")
    return (x >= float(th)).fillna(False).astype(int)

# ============================================================
# #(3) スコア計算（新仕様：補*対応 + defensive導入 + その他製品特別）
#   - defensive点：40点（配当金額ベース defensive比率：0%→0点、50%→40点、50%以上は40点）
#   - defensive判定：L_def_score>=60 を L_def=1 とみなす
# ============================================================
OTHER_PRODUCTS = "その他製品"

def _score_dividend_yield_100(total_invest: float, total_dividend: float) -> tuple[float, float]:
    """利回り[%]とスコア(0-100)を返す。4%→0点、10%→100点（線形、クリップ）"""
    if total_invest <= 0:
        return 0.0, 0.0
    y = (total_dividend / total_invest) * 100.0
    y0, y1 = 4.0, 10.0
    score = (y - y0) / (y1 - y0) * 100.0
    score = max(0.0, min(100.0, float(score)))
    return float(y), float(score)

def _score_industries_by_dividend_share_40(work_df: pd.DataFrame, all_categories_eval: list[str]) -> tuple[float, list[str]]:
    """
    業種スコア(0-40)とメッセージlistを返す
    - 配当金額（補正なし: 配当金額）ベース
    - shareの分母は total_dividend
    - all_categories_eval（ETF除外済み）を軸にする
    - 「その他製品」だけ15%超でも減点しない（満点キープ）＋15%超をメッセージ表示
    """
    msgs: list[str] = []

    total_dividend = float(pd.to_numeric(work_df["配当金額"], errors="coerce").fillna(0.0).sum())
    if total_dividend <= 0:
        return 0.0, msgs

    share = (work_df.groupby("industry_33")["配当金額"].sum() / total_dividend * 100.0)
    share = share.reindex(all_categories_eval, fill_value=0.0)

    n = len(all_categories_eval)
    per_max = 40.0 / n  # 40点を業種数で均等配分

    def piecewise_points_general(r_pct: float) -> float:
        r = float(r_pct)
        if r <= 0.0:
            return 0.0
        if 0.0 < r < 1.5:
            return per_max * (r / 1.5)
        if 1.5 <= r <= 15.0:
            return per_max
        if 15.0 < r < 17.0:
            return per_max * (1.0 - (r - 15.0) / 2.0)
        return 0.0

    def piecewise_points_other_products(r_pct: float) -> float:
        r = float(r_pct)
        if r <= 0.0:
            return 0.0
        if 0.0 < r < 1.5:
            return per_max * (r / 1.5)
        return per_max  # 1.5%以上は満点キープ（15%超でも減点しない）

    if OTHER_PRODUCTS in share.index:
        r_other = float(share.loc[OTHER_PRODUCTS])
        if r_other > 15.0:
            msgs.append(f"注意：業種「{OTHER_PRODUCTS}」の配当比率が {r_other:.1f}%（15%超）です。")

    score = 0.0
    for ind, r in share.items():
        if ind == OTHER_PRODUCTS:
            score += piecewise_points_other_products(r)
        else:
            score += piecewise_points_general(r)

    score = max(0.0, min(40.0, float(score)))
    return float(score), msgs

def _score_stocks_by_invest_concentration_20(work_df: pd.DataFrame, invest_col: str = "投資金額_補") -> float:
    """銘柄集中スコア(0-20)。投資比率(補*)が3%を超えた分[%]を減点"""
    total_invest = float(pd.to_numeric(work_df[invest_col], errors="coerce").fillna(0.0).sum())
    if total_invest <= 0:
        return 0.0

    by_code = work_df.groupby("code")[invest_col].sum()
    weights = by_code / total_invest

    excess = (weights - 0.03).clip(lower=0.0)
    penalty_points = float((excess * 100.0).sum())

    score = 20.0 - penalty_points
    return max(0.0, float(score))

def _score_defensive_by_dividend_share_40(work_df: pd.DataFrame) -> tuple[float, float]:
    """
    defensive比率(%)とスコア(0-40)を返す（配当金額ベース）
    - defensive判定：L_def_score >= 60 を L_def=1 とみなす
    - 0%→0点、50%→40点、50%以上は40点キープ
    """
    total_dividend = float(pd.to_numeric(work_df["配当金額"], errors="coerce").fillna(0.0).sum())
    if total_dividend <= 0:
        return 0.0, 0.0

    defscore = pd.to_numeric(work_df.get("L_def_score", pd.Series([pd.NA] * len(work_df))), errors="coerce")
    ldef_auto = (defscore >= float(DEF_SCORE_TH)).fillna(False)

    def_div = float(pd.to_numeric(work_df.loc[ldef_auto, "配当金額"], errors="coerce").fillna(0.0).sum())

    ratio = (def_div / total_dividend) * 100.0
    score = (ratio / 50.0) * 40.0
    score = max(0.0, min(40.0, float(score)))
    return float(ratio), float(score)

def _calc_portfolio_score(work_df: pd.DataFrame, all_categories_eval: list[str]) -> dict:
    """
    200点満点のスコア辞書を返す
    """
    has_sup = "投資金額_補" in work_df.columns
    invest_col = "投資金額_補" if has_sup else "投資金額"

    total_invest_sup = float(pd.to_numeric(work_df[invest_col], errors="coerce").fillna(0.0).sum())
    total_invest_raw = float(pd.to_numeric(work_df["投資金額"], errors="coerce").fillna(0.0).sum()) if "投資金額" in work_df.columns else total_invest_sup
    total_dividend = float(pd.to_numeric(work_df["配当金額"], errors="coerce").fillna(0.0).sum())

    # 利回り（通常/補*）
    y_raw = (total_dividend / total_invest_raw) * 100.0 if total_invest_raw > 0 else 0.0
    y_sup, score_y = _score_dividend_yield_100(total_invest_sup, total_dividend)

    # 業種
    score_i, msgs = _score_industries_by_dividend_share_40(work_df, all_categories_eval)

    # 銘柄集中（補*ベース）
    score_s = _score_stocks_by_invest_concentration_20(work_df, invest_col="投資金額_補" if has_sup else "投資金額")

    # defensive（配当金額ベース：def_score>=60 を defensive とみなす）
    def_ratio_pct, score_d = _score_defensive_by_dividend_share_40(work_df)

    total_score = score_y + score_i + score_s + score_d

    return {
        "投資金額合計_通常": total_invest_raw,
        "投資金額合計_補": total_invest_sup,
        "年間配当合計": total_dividend,

        "配当利回り(%)": float(y_raw),
        "配当利回り(補*)(%)": float(y_sup),

        "defensive比率(%)": float(def_ratio_pct),

        "スコア_配当利回り(100)": float(score_y),
        "スコア_業種(40)": float(score_i),
        "スコア_銘柄(20)": float(score_s),
        "スコア_def(40)": float(score_d),

        "スコア_合計(200)": float(total_score),
        "messages": msgs,
    }

# ============================================================
# #(3-1) 追加：表示用ユーティリティ（候補一覧のフォーマット）
#   ★200点満点 + Δdef 対応版（この定義を最後にして上書き勝ちさせる）
# ============================================================
def _display_candidates_table(res: pd.DataFrame, topn: int):
    """
    TOP表（200点満点 / Δdefあり）：
      code, name, 業種, L_def_auto, L_def_score,
      株価, 株数, 投資金額,
      配当, 利回り, 利回り(補*),
      Δ総スコア, Δ配当, Δ業種, Δ銘柄, Δdef,
      総スコア
    """
    if res is None or len(res) == 0:
        return

    show = res.head(topn).copy()

    desired_cols = [
        "code", "name", "業種",
        "L_def_auto", "L_def_score",
        "株価", "株数", "投資金額",
        "配当", "利回り", "利回り(補*)",
        "Δ総スコア", "Δ配当", "Δ業種", "Δ銘柄", "Δdef",
        "総スコア",
    ]
    cols = [c for c in desired_cols if c in show.columns]
    show = show[cols]

    fmt = {}

    # defensive表示（自動判定）
    if "L_def_auto" in show.columns:
        fmt["L_def_auto"] = "{:.0f}"
    if "L_def_score" in show.columns:
        fmt["L_def_score"] = "{:.1f}"

    for c in ["株価", "投資金額"]:
        if c in show.columns:
            fmt[c] = "{:,.0f}"
    if "株数" in show.columns:
        fmt["株数"] = "{:,.0f}"

    if "配当" in show.columns:
        fmt["配当"] = "{:.2f}"
    for c in ["利回り", "利回り(補*)"]:
        if c in show.columns:
            fmt[c] = "{:.2f}"

    for c in ["Δ総スコア", "Δ配当", "Δ業種", "Δ銘柄", "Δdef"]:
        if c in show.columns:
            fmt[c] = "{:+.2f}"

    if "総スコア" in show.columns:
        fmt["総スコア"] = "{:.2f}"

    try:
        display(show.style.format(fmt))
    except Exception:
        display(show)
        
# ============================================================
# #(3-2) 追加：売却候補提案（簡易ヒューリスティック）
#   - 目的：スコアが上がる可能性が高い「売却（全売/一部売）」案を出す
#   - 観点：
#       A) 銘柄集中（投資比率>3%）→ 3%まで落とす売却株数（100株単位）
#       B) 業種集中（配当比率>15%）→ 15%まで落とすための売却（配当が大きい銘柄から）
#       C) 低利回り（個別利回り<3.75%）→ 全売候補（投資比率が一定以上のもの）
#   - 各候補は「単独でその売却を行った場合のスコア差分」も計算して並べる
# ============================================================
def _suggest_sell_candidates(
    df_current: pd.DataFrame,
    all_categories_eval: list[str],
    low_yield_floor_pct: float = 3.75,
    stock_weight_cap: float = 0.03,
    industry_div_cap_pct: float = 15.0,
    min_invest_share_for_low_yield: float = 0.01,
) -> pd.DataFrame:
    base = _calc_portfolio_score(df_current, all_categories_eval)

    # codeごとにまとめる（同一銘柄が複数行ある場合に対応）
    g = df_current.groupby(["code", "name", "industry_33"], dropna=False).agg(
        株数=("株数", "sum"),
        投資金額=("投資金額", "sum"),
        配当金額=("配当金額", "sum"),
        年間配当=("年間配当", "mean"),     # 参考
        取得単価=("取得単価", "mean"),     # 参考（売却株数算定の簡易に使用）
    ).reset_index()

    total_inv = float(g["投資金額"].sum())
    total_div = float(g["配当金額"].sum())
    if total_inv <= 0 or total_div <= 0:
        return pd.DataFrame()

    g["投資比率"] = g["投資金額"] / total_inv
    g["個別利回り(%)"] = np.where(g["投資金額"] > 0, (g["配当金額"] / g["投資金額"]) * 100.0, np.nan)

    suggestions = []

    # --------------------------------------------------------
    # A) 銘柄集中（>3%）
    # --------------------------------------------------------
    over = g[g["投資比率"] > stock_weight_cap].copy()
    for _, r in over.iterrows():
        code = str(r["code"])
        inv = float(r["投資金額"])
        shares = float(r["株数"])
        unit_cost = float(r["取得単価"]) if pd.notna(r["取得単価"]) else np.nan
        if not (unit_cost > 0):
            continue

        # 3%まで落とすために必要な投資金額削減
        target_inv = stock_weight_cap * total_inv
        reduce_inv = inv - target_inv
        if reduce_inv <= 0:
            continue

        # 売却株数（100株単位、超えない）
        sell_shares = math.ceil((reduce_inv / unit_cost) / 100.0) * 100
        sell_shares = int(min(sell_shares, math.floor(shares / 100.0) * 100))
        if sell_shares <= 0:
            continue

        # 売却を反映したDFを作る（簡易：対象銘柄の株数を減らす）
        df_new = df_current.copy()
        # code一致行に按分して減らす（投資金額比率の改善が目的なので近似でOK）
        idx = df_new[df_new["code"].astype("string") == code].index
        if len(idx) == 0:
            continue
        # 先頭行にまとめて減らす（簡易）
        i0 = idx[0]
        df_new.at[i0, "株数"] = max(0.0, float(df_new.at[i0, "株数"]) - sell_shares)

        # 再計算
        df_new["投資金額"] = df_new["株数"] * df_new["取得単価"]
        df_new["配当金額"] = df_new["株数"] * df_new["年間配当"]

        info = _calc_portfolio_score(df_new, all_categories_eval)
        suggestions.append({
            "銘柄コード": code,
            "銘柄名": r.get("name", ""),
            "業種": r.get("industry_33", ""),
            "提案": f"一部売却（銘柄集中）",
            "売却株数": sell_shares,
            "売却後想定_投資比率(%)": (inv - sell_shares * unit_cost) / total_inv * 100.0,
            "現状_投資比率(%)": float(r["投資比率"]) * 100.0,
            "現状_個別利回り(%)": float(r["個別利回り(%)"]),
            "差分_スコア合計": info["スコア_合計(100)"] - base["スコア_合計(100)"],
            "差分_配当利回り点": info["スコア_配当利回り(40)"] - base["スコア_配当利回り(40)"],
            "差分_業種点": info["スコア_業種(40)"] - base["スコア_業種(40)"],
            "差分_銘柄点": info["スコア_銘柄(20)"] - base["スコア_銘柄(20)"],
        })

    # --------------------------------------------------------
    # B) 業種集中（配当比率>15%）
    # --------------------------------------------------------
    by_ind = df_current.groupby("industry_33")["配当金額"].sum()
    by_ind = by_ind[by_ind > 0].sort_values(ascending=False)
    for ind, ind_div in by_ind.items():
        if str(ind) in EXCLUDED_INDUSTRIES:
            continue
        share = float(ind_div) / float(total_div) * 100.0
        if share <= industry_div_cap_pct:
            continue

        c = industry_div_cap_pct / 100.0
        need_x = (float(ind_div) - c * float(total_div)) / (1.0 - c)  # x >= ...
        if need_x <= 0:
            continue

        # その業種の銘柄を配当金額が大きい順に並べ、必要分の配当を落とす
        gi = g[g["industry_33"].astype("string") == str(ind)].copy()
        gi = gi.sort_values("配当金額", ascending=False)

        remain = need_x
        for _, r in gi.iterrows():
            code = str(r["code"])
            annual_div_per_share = float(r["年間配当"]) if pd.notna(r["年間配当"]) else np.nan
            if not (annual_div_per_share > 0):
                continue
            shares = float(r["株数"])
            max_reduce_div = float(r["配当金額"])
            if max_reduce_div <= 0:
                continue

            take = min(remain, max_reduce_div)
            # 配当takeを落とすのに必要な株数（100株単位）
            sell_shares = math.ceil((take / annual_div_per_share) / 100.0) * 100
            sell_shares = int(min(sell_shares, math.floor(shares / 100.0) * 100))
            if sell_shares <= 0:
                continue

            # シミュレーション（簡易：対象銘柄の先頭行から減らす）
            df_new = df_current.copy()
            idx = df_new[df_new["code"].astype("string") == code].index
            if len(idx) == 0:
                continue
            i0 = idx[0]
            df_new.at[i0, "株数"] = max(0.0, float(df_new.at[i0, "株数"]) - sell_shares)

            df_new["投資金額"] = df_new["株数"] * df_new["取得単価"]
            df_new["配当金額"] = df_new["株数"] * df_new["年間配当"]

            info = _calc_portfolio_score(df_new, all_categories_eval)
            suggestions.append({
                "銘柄コード": code,
                "銘柄名": r.get("name", ""),
                "業種": str(ind),
                "提案": f"一部売却（業種集中：{share:.1f}%→{industry_div_cap_pct:.1f}%狙い）",
                "売却株数": sell_shares,
                "売却後想定_投資比率(%)": np.nan,
                "現状_投資比率(%)": float(r["投資比率"]) * 100.0 if "投資比率" in r else np.nan,
                "現状_個別利回り(%)": float(r["個別利回り(%)"]) if "個別利回り(%)" in r else np.nan,
                "差分_スコア合計": info["スコア_合計(100)"] - base["スコア_合計(100)"],
                "差分_配当利回り点": info["スコア_配当利回り(40)"] - base["スコア_配当利回り(40)"],
                "差分_業種点": info["スコア_業種(40)"] - base["スコア_業種(40)"],
                "差分_銘柄点": info["スコア_銘柄(20)"] - base["スコア_銘柄(20)"],
            })

            remain -= sell_shares * annual_div_per_share
            if remain <= 0:
                break

    # --------------------------------------------------------
    # C) 低利回り（個別利回り<3.75%）→ 全売候補（投資比率が一定以上）
    # --------------------------------------------------------
    low = g[(g["個別利回り(%)"] < low_yield_floor_pct) & (g["投資比率"] >= min_invest_share_for_low_yield)].copy()
    for _, r in low.iterrows():
        code = str(r["code"])
        shares = int(math.floor(float(r["株数"]) / 100.0) * 100)
        if shares <= 0:
            continue

        df_new = df_current.copy()
        idx = df_new[df_new["code"].astype("string") == code].index
        if len(idx) == 0:
            continue
        i0 = idx[0]
        df_new.at[i0, "株数"] = max(0.0, float(df_new.at[i0, "株数"]) - shares)

        df_new["投資金額"] = df_new["株数"] * df_new["取得単価"]
        df_new["配当金額"] = df_new["株数"] * df_new["年間配当"]

        info = _calc_portfolio_score(df_new, all_categories_eval)
        suggestions.append({
            "銘柄コード": code,
            "銘柄名": r.get("name", ""),
            "業種": r.get("industry_33", ""),
            "提案": f"全売候補（低利回り<{low_yield_floor_pct:.2f}%）",
            "売却株数": shares,
            "売却後想定_投資比率(%)": np.nan,
            "現状_投資比率(%)": float(r["投資比率"]) * 100.0,
            "現状_個別利回り(%)": float(r["個別利回り(%)"]),
            "差分_スコア合計": info["スコア_合計(100)"] - base["スコア_合計(100)"],
            "差分_配当利回り点": info["スコア_配当利回り(40)"] - base["スコア_配当利回り(40)"],
            "差分_業種点": info["スコア_業種(40)"] - base["スコア_業種(40)"],
            "差分_銘柄点": info["スコア_銘柄(20)"] - base["スコア_銘柄(20)"],
        })

    out = pd.DataFrame(suggestions)
    if len(out) == 0:
        return out

    # スコア上昇が大きい順
    out = out.sort_values("差分_スコア合計", ascending=False).reset_index(drop=True)
    return out


# ============================================================
# #(4) 配当採用ロジック（候補銘柄の年間配当を作る）
#   - 予があれば最新年度の予、なければ最新年度の実績
# ============================================================
def _build_haitou_pick_df() -> pd.DataFrame:
    haitou_df = pd.read_csv("Data/03_haitou.csv", dtype={"code": "string"}, encoding="utf-8-sig")
    if "code" not in haitou_df.columns:
        for c in haitou_df.columns:
            if str(c).replace("\ufeff", "") == "code":
                haitou_df = haitou_df.rename(columns={c: "code"})
                break
    if "code" not in haitou_df.columns:
        raise KeyError("03_haitou.csv に code 列が見つかりません。")

    haitou_df["code"] = normalize_code_str_series(haitou_df["code"])

    year_cols = [c for c in haitou_df.columns if re.fullmatch(r"\d{4}", str(c))]
    year_cols_sorted = sorted(year_cols, key=lambda x: int(x))

    def _parse_div_cell(x):
        if pd.isna(x):
            return (False, None)
        s = str(x).strip()
        if s == "":
            return (False, None)
        is_forecast = "予" in s
        m = re.search(r"(\d+(?:\.\d+)?)", s)
        if not m:
            return (is_forecast, None)
        return (is_forecast, float(m.group(1)))

    def _pick_latest_dividend(row):
        latest_forecast = (None, None)
        latest_actual = (None, None)
        for y in year_cols_sorted:
            is_fc, val = _parse_div_cell(row.get(y))
            if val is None:
                continue
            yi = int(y)
            if is_fc:
                if latest_forecast[0] is None or yi > latest_forecast[0]:
                    latest_forecast = (yi, val)
            else:
                if latest_actual[0] is None or yi > latest_actual[0]:
                    latest_actual = (yi, val)

        if latest_forecast[0] is not None:
            return pd.Series({"年間配当": latest_forecast[1], "年間配当_年度": latest_forecast[0], "年間配当_区分": "予想"})
        if latest_actual[0] is not None:
            return pd.Series({"年間配当": latest_actual[1], "年間配当_年度": latest_actual[0], "年間配当_区分": "実績"})
        return pd.Series({"年間配当": pd.NA, "年間配当_年度": pd.NA, "年間配当_区分": pd.NA})

    picked = haitou_df.apply(_pick_latest_dividend, axis=1)
    haitou_pick_df = pd.concat([haitou_df[["code"]].copy(), picked], axis=1)
    haitou_pick_df = (
        haitou_pick_df
        .sort_values(["code", "年間配当_年度"])
        .drop_duplicates(subset=["code"], keep="last")
    )
    return haitou_pick_df

# ============================================================
# #(X-1) summary.csv（2行目）から「総合実現損益」を辞書化
#   - key: 4桁コード（文字列） or 銘柄名
#   - value: 実現損益（float）
#   - ここでは code優先で引けるように「code用map」を返す（銘柄名は補助）
# ============================================================
def _load_realized_pl_map_summary2() -> dict[str, float]:
    summary_path = "Data/jyouto/summary.csv"
    if not os.path.exists(summary_path):
        return {}

    summary_df = pd.read_csv(summary_path, dtype="string", encoding="utf-8-sig")
    if len(summary_df) < 2:
        return {}

    row2 = summary_df.iloc[1]

    pl_map_by_code: dict[str, float] = {}
    pl_map_by_name: dict[str, float] = {}

    for k, v in row2.items():
        if k is None:
            continue
        key_raw = str(k).strip()
        if key_raw == "":
            continue

        val_num = _to_num_scalar(v)
        if pd.isna(val_num):
            continue

        key_norm = (
            pd.Series([key_raw], dtype="string")
              .pipe(normalize_code_str_series)
              .iloc[0]
        )
        if re.fullmatch(r"\d{4}", str(key_norm)):
            pl_map_by_code[str(key_norm)] = float(val_num)
        else:
            pl_map_by_name[key_raw] = float(val_num)

    # 今回は propose_recommendations 側で code.map() する想定なので code-map を返す
    return pl_map_by_code


# ============================================================
# #(X-2) スコア計算（補*対応版）
#   - 配当利回り：投資金額は「投資金額_補」を優先（なければ投資金額）
#   - 銘柄集中：投資比率は「投資金額_補」で計算
#   - 業種：配当金額ベース（従来通り）
# ============================================================
def _score_dividend_yield(total_invest: float, total_dividend: float) -> tuple[float, float]:
    """
    配当利回り[%]を算出し、
      - 3.75% → 0点
      - 50%   → 40点
    の線形（上下限クリップ）で採点する。
    """
    if total_invest <= 0:
        return 0.0, 0.0

    y = (total_dividend / total_invest) * 100.0

    y0 = 3.75
    y1 = 50.0
    if y <= y0:
        score = 0.0
    elif y >= y1:
        score = 40.0
    else:
        score = (y - y0) / (y1 - y0) * 40.0

    return float(y), float(score)

# ============================================================
# #(5) メイン関数：推奨銘柄提案（新スコア + Δdef 対応）
#   ★変更点（今回の要件）：
#     - defensive判定は L_def_score>=60 で自動生成（L_defは使わない）
#     - df_current / cand_df に L_def_score を取り込み、必要なら L_def_auto も作る（表示用）
# ============================================================
def propose_recommendations(
    owner: str | None = None,
    topn: int = 30,
    include_owned_as_candidate: bool = False,
    restrict_candidate_years_to_this_and_last: bool = True,
):
    # ========================================================
    # (A) 現状ポートフォリオ作成（target_dfベース） + 補*用の投資金額_補
    # ========================================================
    df_current = target_df.copy()
    df_current["code"] = normalize_code_str_series(df_current["code"])

    if owner is not None:
        df_current = df_current[df_current["所有者"] == owner].copy()

    df_current["株数"] = pd.to_numeric(df_current["株数"], errors="coerce")
    df_current["取得単価"] = pd.to_numeric(df_current["取得単価"], errors="coerce")
    df_current["年間配当"] = pd.to_numeric(df_current["年間配当"], errors="coerce")

    df_current["投資金額"] = df_current["株数"] * df_current["取得単価"]
    df_current["配当金額"] = df_current["株数"] * df_current["年間配当"]

    valid_industry = (
        df_current["industry_33"].notna()
        & ~df_current["industry_33"].astype("string").str.strip().isin(["", "-", "0"])
    )
    df_current = df_current[valid_industry].copy()

    if len(df_current) == 0:
        raise ValueError("現状ポートフォリオが空になりました。target_df / owner指定を確認してください。")

    # --- 補*：実現益（Tのみ・正のみ）を反映して投資金額_補 を作る ---
    pl_map = _load_realized_pl_map_summary2()
    df_current["実現損益"] = df_current["code"].astype("string").map(pl_map).fillna(0.0)
    df_current["実現益_反映"] = np.where(
        (df_current["所有者"].astype("string") == "T") & (df_current["実現損益"] > 0),
        df_current["実現損益"],
        0.0
    )
    df_current["投資金額_補"] = df_current["投資金額"] - df_current["実現益_反映"]

    title = "全体" if owner is None else f"所有者={owner}"

    # ========================================================
    # (B) IDmap読み込み（カテゴリ／候補抽出） + L_def_score を取り込む
    # ========================================================
    idmap_df = pd.read_csv("Data/01_IDmap.csv", dtype={"code": "string"}, encoding="utf-8-sig")
    if "code" not in idmap_df.columns:
        for c in idmap_df.columns:
            if str(c).replace("\ufeff", "") == "code":
                idmap_df = idmap_df.rename(columns={c: "code"})
                break
    if "code" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に code 列が見つかりません。")
    if "industry_33" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に industry_33 列が見つかりません。")
    if "L_高配当" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に L_高配当 列が見つかりません。")

    # ★ L_def_score は必須（def_score思想のため）。無いならエラーにする
    if "L_def_score" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に L_def_score 列が見つかりません（def_scoreを保存している列）。")

    idmap_df["code"] = normalize_code_str_series(idmap_df["code"])

    # df_current 側に L_def_score があれば衝突回避で一旦消す
    if "L_def_score" in df_current.columns:
        df_current = df_current.drop(columns=["L_def_score"])

    # 現状ポートフォリオにも L_def_score を付与（defensive比率計算用）
    df_current = pd.merge(
        df_current,
        idmap_df[["code", "L_def_score"]].copy(),
        on="code",
        how="left",
    )
    df_current["L_def_score"] = pd.to_numeric(df_current["L_def_score"], errors="coerce")

    # （任意）表示・レビュー用：自動L_def
    df_current["L_def_auto"] = _ldef_from_defscore_series(df_current["L_def_score"])

    # 業種一覧（ETF2業種は評価から除外）
    all_categories = (
        idmap_df["industry_33"]
        .astype("string")
        .map(lambda v: v.strip() if isinstance(v, str) else v)
        .dropna()
    )
    all_categories = sorted({v for v in all_categories.unique().tolist() if v not in ["", "-", "0"]})
    all_categories_eval = [c for c in all_categories if c not in EXCLUDED_INDUSTRIES]
    if len(all_categories_eval) == 0:
        raise ValueError("評価対象の industry_33 が 0 件になりました。除外設定やIDmapを確認してください。")

    # 候補銘柄（高配当ラベル=1）
    cand_df = idmap_df.copy()
    cand_df["L_高配当_num"] = pd.to_numeric(cand_df["L_高配当"], errors="coerce").fillna(0).astype(int)
    cand_df = cand_df[cand_df["L_高配当_num"] == 1].copy()

    # 候補にも数値化した L_def_score を持たせる
    cand_df["L_def_score"] = pd.to_numeric(cand_df["L_def_score"], errors="coerce")
    cand_df["L_def_auto"] = _ldef_from_defscore_series(cand_df["L_def_score"])

    # 候補に配当を付与
    haitou_pick_df = _build_haitou_pick_df()
    cand_df = pd.merge(cand_df, haitou_pick_df, on="code", how="left")
    cand_df["年間配当"] = pd.to_numeric(cand_df["年間配当"], errors="coerce")

    if restrict_candidate_years_to_this_and_last:
        current_year = pd.Timestamp.today().year
        target_years = {current_year, current_year - 1}
        cand_df = cand_df[
            cand_df["年間配当"].notna()
            & cand_df["年間配当_年度"].notna()
            & cand_df["年間配当_年度"].astype("Int64").isin(target_years)
        ].copy()
    else:
        cand_df = cand_df[cand_df["年間配当"].notna()].copy()

    # ========================================================
    # (C) 株価（最新kabuka）／現状スコア（200点）
    # ========================================================
    kabuka_path = _find_latest_kabuka_file()
    kabuka_df = _read_kabuka_wide(kabuka_path)

    base = _calc_portfolio_score(df_current, all_categories_eval)

    base_summary_lines = [
        f"==== 現状ポートフォリオ（{title}） ====",
        f"株価参照: {os.path.basename(kabuka_path)}",
        f"スコア合計: {base['スコア_合計(200)']:.2f} / 200",
        f"  配当利回り: {base['スコア_配当利回り(100)']:.2f} / 100（利回り(補*) {base['配当利回り(補*)(%)']:.2f}%）",
        f"  業種      : {base['スコア_業種(40)']:.2f} / 40",
        f"  銘柄      : {base['スコア_銘柄(20)']:.2f} / 20",
        f"  defensive : {base['スコア_def(40)']:.2f} / 40（def比率 {base['defensive比率(%)']:.1f}%）",
        f"投資金額合計(補*): {base['投資金額合計_補']:,.0f}円 / 年間配当合計: {base['年間配当合計']:,.0f}円",
    ]
    for m in base.get("messages", []):
        base_summary_lines.append(m)
    base_summary_lines.append("")

    # ========================================================
    # (D) 減点対象の抽出（銘柄集中/業種集中）
    #   ★表示用：g_stock に L_def_score と L_def_auto を含める
    # ========================================================
    g_stock = df_current.groupby(["code", "name", "industry_33"], dropna=False).agg(
        株数=("株数", "sum"),
        投資金額_補=("投資金額_補", "sum"),
        投資金額=("投資金額", "sum"),
        取得単価=("取得単価", "mean"),
        配当金額=("配当金額", "sum"),
        L_def_score=("L_def_score", "max"),
        L_def_auto=("L_def_auto", "max"),
    ).reset_index()

    total_inv_sup = float(pd.to_numeric(g_stock["投資金額_補"], errors="coerce").fillna(0.0).sum())
    if total_inv_sup > 0:
        g_stock["投資割合(補*)(%)"] = g_stock["投資金額_補"] / total_inv_sup * 100.0
        g_stock["超過(>3%)(%)"] = (g_stock["投資割合(補*)(%)"] - 3.0).clip(lower=0.0)
        g_stock["減点(点)"] = g_stock["超過(>3%)(%)"]
    else:
        g_stock["投資割合(補*)(%)"] = np.nan
        g_stock["超過(>3%)(%)"] = 0.0
        g_stock["減点(点)"] = 0.0

    penal_stock = g_stock[g_stock["超過(>3%)(%)"] > 0].copy()
    if len(penal_stock) > 0:
        target_inv = 0.03 * total_inv_sup
        penal_stock["減点解消_目標株数"] = np.where(
            (penal_stock["取得単価"].notna()) & (penal_stock["取得単価"] > 0),
            np.floor((target_inv / penal_stock["取得単価"]) / 100.0) * 100.0,
            np.nan
        )
        penal_stock = penal_stock.sort_values(["減点(点)", "投資割合(補*)(%)"], ascending=[False, False])

    by_ind = df_current.groupby("industry_33")["配当金額"].sum()
    total_div = float(by_ind.sum())
    ind_share = ((by_ind / total_div) * 100.0) if total_div > 0 else pd.Series(dtype=float)
    ind_share = ind_share.dropna().sort_values(ascending=False)

    penal_inds = ind_share[(ind_share > 15.0) & (~ind_share.index.astype("string").isin(EXCLUDED_INDUSTRIES))].copy()
    penal_ind_rows = []
    if len(penal_inds) > 0:
        g_stock2 = g_stock.copy()
        g_stock2["個別利回り(補*)(%)"] = np.where(
            g_stock2["投資金額_補"] > 0,
            (g_stock2["配当金額"] / g_stock2["投資金額_補"]) * 100.0,
            np.nan
        )
        for ind, share_pct in penal_inds.items():
            gi = g_stock2[g_stock2["industry_33"].astype("string") == str(ind)].copy()
            if len(gi) == 0:
                continue
            top_inv = gi.sort_values("投資金額_補", ascending=False).iloc[0]
            gi_y = gi[gi["個別利回り(補*)(%)"].notna()].copy()
            low_y = gi_y.sort_values("個別利回り(補*)(%)", ascending=True).iloc[0] if len(gi_y) > 0 else top_inv
            penal_ind_rows.append({
                "業種": str(ind),
                "配当比率(%)": float(share_pct),
                "代表_投資最大": f"{top_inv['code']} {top_inv.get('name','')}",
                "代表_低利回り": f"{low_y['code']} {low_y.get('name','')}",
            })

    # ========================================================
    # (E) 候補の絞り（既保有除外 or 買い増し許可）
    # ========================================================
    owned_codes = set(df_current["code"].astype("string").dropna().tolist())
    if not include_owned_as_candidate:
        cand_eval = cand_df[~cand_df["code"].astype("string").isin(owned_codes)].copy()
    else:
        cand_eval = cand_df.copy()

    # ========================================================
    # (F) 1銘柄追加の評価 → res（L_def_auto / L_def_score を持たせる）
    # ========================================================
    rows = []
    for _, r in cand_eval.iterrows():
        code = str(r["code"])
        name = r.get("name", "")
        industry = r.get("industry_33", None)
        annual_div = r.get("年間配当", np.nan)
        def_score = r.get("L_def_score", np.nan)
        l_def_auto = int(r.get("L_def_auto", 0))

        if industry is None or pd.isna(industry) or str(industry).strip() in ["", "-", "0"]:
            continue
        industry = str(industry).strip()
        if industry in EXCLUDED_INDUSTRIES:
            continue

        price = _latest_price_from_kabuka(kabuka_df, code)
        if np.isnan(price) or price <= 0:
            continue

        buy_shares = _calc_buy_shares(price)
        if buy_shares <= 0:
            continue

        buy_invest = float(buy_shares * price)
        annual_div = float(annual_div) if (annual_div is not None and not pd.isna(annual_div)) else 0.0
        annual_div_amount = float(buy_shares * annual_div)

        cand_yield_pct = (annual_div / price) * 100.0 if (price > 0) else np.nan
        cand_yield_sup_pct = cand_yield_pct

        add_row = {
            "code": code,
            "industry_33": industry,
            "株数": float(buy_shares),
            "取得単価": float(price),
            "年間配当": float(annual_div),
            "投資金額": buy_invest,
            "配当金額": annual_div_amount,
            "所有者": "ADD",
            "name": name,
            "実現益_反映": 0.0,
            "投資金額_補": buy_invest,
            "L_def_score": def_score,   # ★ defensive比率計算に使う
            "L_def_auto": l_def_auto,   # ★ 表示・レビュー用
        }

        df_new = pd.concat([df_current, pd.DataFrame([add_row])], ignore_index=True)
        info = _calc_portfolio_score(df_new, all_categories_eval)

        rows.append({
            "code": code,
            "name": name,
            "業種": industry,
            "L_def_auto": l_def_auto,
            "L_def_score": def_score,

            "株価": float(price),
            "株数": int(buy_shares),
            "投資金額": float(buy_invest),

            "配当": float(annual_div),
            "利回り": float(cand_yield_pct),
            "利回り(補*)": float(cand_yield_sup_pct),

            "Δ総スコア": info["スコア_合計(200)"] - base["スコア_合計(200)"],
            "Δ配当": info["スコア_配当利回り(100)"] - base["スコア_配当利回り(100)"],
            "Δ業種": info["スコア_業種(40)"] - base["スコア_業種(40)"],
            "Δ銘柄": info["スコア_銘柄(20)"] - base["スコア_銘柄(20)"],
            "Δdef": info["スコア_def(40)"] - base["スコア_def(40)"],

            "総スコア": info["スコア_合計(200)"],
        })

    res = pd.DataFrame(rows)
    if len(res) == 0:
        print("候補が0件になりました。以下を確認してください：")
        print("- 01_IDmap.csv の L_高配当==1 が存在するか")
        print("- 03_haitou.csv に候補銘柄の配当が入っているか（今年or昨年縛りを外すなら restrict_candidate_years_to_this_and_last=False）")
        print("- Data/kabuka/YYYY.csv に候補銘柄コード列が存在し、終値が入っているか")
        return res

    res = res.sort_values(["Δ総スコア", "投資金額"], ascending=[False, True]).reset_index(drop=True)

    # ========================================================
    # (G-0) 推奨Top10 codeリスト
    # ========================================================
    top10_codes = (
        res["code"]
        .astype("string")
        .dropna()
        .head(10)
        .astype(str)
        .tolist()
    )
    quoted = ", ".join([f"'{c}'" for c in top10_codes])
    list_text = f"[{quoted}]"

    print("\n" + "=" * 100)
    print(f"推奨Top10 codeリスト（{title}）※コピペ用")
    print("=" * 100)
    print(list_text)
    print(f"detail({list_text})")
    print()

    for line in base_summary_lines:
        print(line)

    # ========================================================
    # (G-1) 減点対象（銘柄/業種）表示 → TOP表の上
    # ========================================================
    if len(penal_stock) > 0:
        print("---- 減点対象（銘柄集中：投資割合>3%） ----")
        show_cols = ["code", "name", "industry_33", "L_def_auto", "L_def_score", "投資割合(補*)(%)", "超過(>3%)(%)", "減点(点)", "株数", "減点解消_目標株数"]
        out = penal_stock[show_cols].copy().rename(columns={"industry_33": "業種"})
        fmt = {
            "L_def_auto": "{:.0f}",
            "L_def_score": "{:.1f}",
            "投資割合(補*)(%)": "{:.2f}",
            "超過(>3%)(%)": "{:.2f}",
            "減点(点)": "{:.2f}",
            "株数": "{:,.0f}",
            "減点解消_目標株数": "{:,.0f}",
        }
        try:
            display(out.style.format(fmt))
        except Exception:
            display(out)
        print()

    if len(penal_ind_rows) > 0:
        print("---- 減点対象（業種集中：配当比率>15%） ----")
        out = pd.DataFrame(penal_ind_rows).sort_values("配当比率(%)", ascending=False)
        try:
            display(out.style.format({"配当比率(%)": "{:.2f}"}))
        except Exception:
            display(out)
        print()

    # ========================================================
    # (G-2) 候補一覧（L_def_auto / L_def_score を含める）
    # ========================================================
    print(f"==== 推奨（{title}：1銘柄追加でスコア最大化）Top{min(topn, len(res))} ====")
    _display_candidates_table(res, topn=topn)
    print()

    return res

# ============================================================
# #(6) 実行例（必要ならコメントアウト）
# ============================================================
# propose_recommendations()      # 全体
# propose_recommendations("Y")   # 所有者Y
# propose_recommendations("T")   # 所有者T
propose_recommendations("T")
print("\n\n**********************************\n\n")
propose_recommendations("Y")
print("\n\n**********************************\n\n")
# propose_recommendations()
# print("\n\n**********************************\n\n")
```

```python
# ============================================================
# #(4-1) 銘柄詳細：detail(["1301","1305"], "T") のように複数コードを一括表示（整理版・単一定義）
#   前提（上から順番に実行する想定）：
#     - target_df が存在（2つ目セルで作成済み）
#     - 3つ目セルで定義済みの関数が存在：
#         _build_haitou_pick_df
#         _find_latest_kabuka_file, _read_kabuka_wide, _latest_price_from_kabuka, _calc_buy_shares
#         _calc_portfolio_score
#     - EXCLUDED_INDUSTRIES, normalize_code_str_series, _to_num_scalar が（2/3つ目により）存在している想定
#
#   補*：
#     - owner == "T" のときのみ
#     - 実現損益 > 0 のときだけ投資金額を減算（損失は無視）
# ============================================================

import os
import re
import glob
from typing import List, Dict, Tuple, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# #(4-2) 4つ目セル内専用：コード正規化（lookup用）
#   - 全角→半角
#   - strip
#   - 数字のみで長さ<=4 は左ゼロ埋めで4桁化（kabuka列/IDmap/haitouに合わせやすく）
#   - 5桁以上（ETF等）はそのまま
# ============================================================
def _norm_code4(x: str) -> str:
    s = str(x).strip().translate(str.maketrans("０１２３４５６７８９", "0123456789"))
    if re.fullmatch(r"\d+", s) and len(s) <= 4:
        return s.zfill(4)
    return s

def _norm_codes4(codes: List[str]) -> List[str]:
    return [_norm_code4(c) for c in codes]


# ============================================================
# #(4-3) 依存チェック（足りないときはここで分かりやすく止める）
# ============================================================
_required_globals = [
    "target_df",
    "_build_haitou_pick_df",
    "_find_latest_kabuka_file", "_read_kabuka_wide", "_latest_price_from_kabuka", "_calc_buy_shares",
    "_calc_portfolio_score",
]
_missing = [name for name in _required_globals if name not in globals()]
if _missing:
    raise NameError(
        "4つ目セルの前提が不足しています。上から順に実行できているか確認してください。不足: "
        + ", ".join(_missing)
    )

# EXCLUDED_INDUSTRIES が無い場合だけ保険
if "EXCLUDED_INDUSTRIES" not in globals():
    EXCLUDED_INDUSTRIES = {"日経ETF", "米国ETF"}  # fallback


# ============================================================
# #(4-4) マスタ作成：code -> (name, industry_33, 年間配当)
#   - IDmap + haitou_pick を結合して辞書化（detailで繰り返し読まない）
# ============================================================
def _build_candidate_master_maps_detail() -> Tuple[Dict[str, str], Dict[str, str], Dict[str, float]]:
    idmap_df = pd.read_csv("Data/01_IDmap.csv", dtype={"code": "string"}, encoding="utf-8-sig")
    # BOM対策
    if "code" not in idmap_df.columns:
        for c in idmap_df.columns:
            if str(c).replace("\ufeff", "") == "code":
                idmap_df = idmap_df.rename(columns={c: "code"})
                break
    if "code" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に code 列が見つかりません。")
    if "industry_33" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に industry_33 列が見つかりません。")
    if "name" not in idmap_df.columns:
        idmap_df["name"] = ""

    idmap_df["code"] = idmap_df["code"].astype("string").map(lambda v: _norm_code4(v))

    haitou_pick_df = _build_haitou_pick_df()
    haitou_pick_df["code"] = haitou_pick_df["code"].astype("string").map(lambda v: _norm_code4(v))
    haitou_pick_df["年間配当"] = pd.to_numeric(haitou_pick_df["年間配当"], errors="coerce").fillna(0.0)

    m = pd.merge(
        idmap_df[["code", "name", "industry_33"]],
        haitou_pick_df[["code", "年間配当"]],
        on="code",
        how="left",
    )
    m["年間配当"] = pd.to_numeric(m["年間配当"], errors="coerce").fillna(0.0)

    code_to_name = dict(zip(m["code"].astype(str), m["name"].fillna("").astype(str)))
    code_to_ind = dict(zip(m["code"].astype(str), m["industry_33"].fillna("").astype(str)))
    code_to_div = dict(zip(m["code"].astype(str), m["年間配当"].astype(float)))

    return code_to_name, code_to_ind, code_to_div


# ============================================================
# #(4-5) summary.csv（2行目）読み込み：実現損益map（1回だけ）
#   - key: 4桁コード or 銘柄名
# ============================================================
def _load_realized_pl_maps_once_detail() -> Tuple[Dict[str, float], Dict[str, float]]:
    summary_path = os.path.join("Data", "jyouto", "summary.csv")
    if not os.path.exists(summary_path):
        return {}, {}

    sdf = pd.read_csv(summary_path, dtype="string", encoding="utf-8-sig")
    if len(sdf) < 2:
        return {}, {}

    row2 = sdf.iloc[1]
    pl_by_code: Dict[str, float] = {}
    pl_by_name: Dict[str, float] = {}

    # _to_num_scalar が無い場合の保険
    def _to_num_scalar_local(x) -> float:
        if pd.isna(x):
            return float("nan")
        s = str(x).replace(",", "").strip()
        m = re.search(r"-?\d+(?:\.\d+)?", s)
        if not m:
            return float("nan")
        try:
            return float(m.group(0))
        except Exception:
            return float("nan")

    to_num = _to_num_scalar if ("_to_num_scalar" in globals()) else _to_num_scalar_local

    for k, v in row2.items():
        if k is None:
            continue
        key_raw = str(k).strip()
        if key_raw == "":
            continue

        val = to_num(v)
        if pd.isna(val):
            continue

        key_norm = _norm_code4(key_raw)
        if re.fullmatch(r"\d{4}", key_norm):
            pl_by_code[key_norm] = float(val)
        else:
            pl_by_name[key_raw] = float(val)

    return pl_by_code, pl_by_name

def _get_realized_pl(code4: str, name: str, pl_by_code: Dict[str, float], pl_by_name: Dict[str, float]) -> float:
    if code4 in pl_by_code:
        return float(pl_by_code[code4])
    if name in pl_by_name:
        return float(pl_by_name[name])
    return 0.0


# ============================================================
# #(4-6) スコア用カテゴリ（industry_33 全種類：ETF2業種除外）
# ============================================================
def _load_all_categories_eval_detail() -> List[str]:
    idmap_df = pd.read_csv("Data/01_IDmap.csv", dtype={"code": "string"}, encoding="utf-8-sig")
    if "industry_33" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に industry_33 列が見つかりません。")

    cats = (
        idmap_df["industry_33"]
        .astype("string")
        .map(lambda v: v.strip() if isinstance(v, str) else v)
        .dropna()
    )
    cats = sorted({v for v in cats.unique().tolist() if v not in ["", "-", "0"]})
    cats_eval = [c for c in cats if c not in set(EXCLUDED_INDUSTRIES)]
    if len(cats_eval) == 0:
        raise ValueError("評価対象の industry_33 が 0 件になりました。除外設定やIDmapを確認してください。")
    return cats_eval


# ============================================================
# #(4-7) 株価時系列（wide）：必要銘柄だけを「CSVをファイルごとに1回」読んで集約
# ============================================================
def _read_kabuka_ts_wide_for_codes_detail(
    codes4: List[str],
    start_date: Optional[pd.Timestamp] = None,
    end_date: Optional[pd.Timestamp] = None,
) -> pd.DataFrame:
    if not codes4:
        return pd.DataFrame(columns=["date"])

    today = pd.Timestamp.today().normalize()
    if start_date is None:
        start_date = pd.Timestamp(year=today.year - 2, month=1, day=1)
    if end_date is None:
        end_date = today

    base_dir = os.path.join("Data", "kabuka")
    paths = sorted(glob.glob(os.path.join(base_dir, "[0-9][0-9][0-9][0-9].csv")))
    if not paths:
        return pd.DataFrame(columns=["date"] + codes4)

    chunks = []
    for p in paths:
        kdf = pd.read_csv(p, encoding="utf-8-sig")
        first_col = kdf.columns[0]
        idx_col = "idx"
        if first_col != idx_col:
            kdf = kdf.rename(columns={first_col: idx_col})

        present = [c for c in codes4 if c in kdf.columns]
        if not present:
            continue

        mask = kdf[idx_col].astype("string").str.match(r"^\d{4}-\d{2}-\d{2}$", na=False)
        tmp = kdf.loc[mask, [idx_col] + present].copy()
        tmp[idx_col] = pd.to_datetime(tmp[idx_col], errors="coerce")
        tmp = tmp.dropna(subset=[idx_col])
        if len(tmp) == 0:
            continue

        tmp = tmp[(tmp[idx_col] >= start_date) & (tmp[idx_col] <= end_date)]
        if len(tmp) == 0:
            continue

        for c in present:
            tmp[c] = pd.to_numeric(tmp[c], errors="coerce")

        tmp = tmp.rename(columns={idx_col: "date"})
        chunks.append(tmp)

    if not chunks:
        return pd.DataFrame(columns=["date"] + codes4)

    out = pd.concat(chunks, ignore_index=True)
    out = out.sort_values("date").drop_duplicates(subset=["date"], keep="last").reset_index(drop=True)

    for c in codes4:
        if c not in out.columns:
            out[c] = np.nan

    return out[["date"] + codes4]


def _plot_price_multi_detail(
    ts_wide: pd.DataFrame,
    codes4: List[str],
    code_to_name: Dict[str, str],
    title: str,
):
    if ts_wide is None or len(ts_wide) == 0 or "date" not in ts_wide.columns:
        print("株価データが見つかりません（Data/kabuka のCSV/列名を確認）。")
        return

    today = pd.Timestamp.today().normalize()
    base_date = pd.Timestamp(year=today.year - 2, month=1, day=1)

    plt.figure(figsize=(12, 5))
    plotted = 0

    for code in codes4:
        if code not in ts_wide.columns:
            continue
        s = ts_wide[["date", code]].copy().dropna(subset=["date"]).sort_values("date")
        s[code] = pd.to_numeric(s[code], errors="coerce")
        s = s.dropna(subset=[code])
        if len(s) == 0:
            continue

        s_after = s[s["date"] >= base_date]
        if len(s_after) == 0:
            continue
        base_val = float(s_after[code].iloc[0])
        if not (base_val > 0):
            continue

        y = s_after[code] / base_val
        label = f"{code} {code_to_name.get(code, '')}".strip()
        plt.plot(s_after["date"], y, label=label)
        plotted += 1

    if plotted == 0:
        print("株価チャート：対象期間にデータがありません（列名/期間/日付行を確認）。")
        return

    plt.title(title)
    plt.xlabel("日付")
    plt.ylabel("指数（基準=今年-2年の1/1を1.0）")
    plt.legend(loc="upper center", bbox_to_anchor=(0.5, -0.18), ncol=3, fontsize=8, frameon=False)
    plt.subplots_adjust(bottom=0.28)
    plt.tight_layout()
    plt.show()


# ============================================================
# #(4-8) detail 本体（単一定義）
#   - codes: 推奨コードのlist（3つ目で出した Top10 をコピペ想定）
#   - owner: "T" or "Y"
# ============================================================
def detail(codes: List[str], owner: str):
    if not isinstance(codes, list) or len(codes) == 0:
        raise ValueError("detail は detail(['1301','1305'], 'T') のように list で銘柄コードを指定してください。")
    if owner is None or str(owner).strip() == "":
        raise ValueError("detail の owner は 'T' のように指定してください。")

    owner = str(owner).strip()

    owners_all = sorted(target_df["所有者"].astype("string").dropna().unique().tolist())
    if owner not in owners_all:
        raise ValueError(f"owner='{owner}' が target_df に見つかりません。存在する所有者: {owners_all}")

    codes4 = _norm_codes4([str(c) for c in codes])

    # マスタ/実現損益/カテゴリ
    code_to_name, code_to_ind, code_to_divps = _build_candidate_master_maps_detail()
    pl_by_code, pl_by_name = _load_realized_pl_maps_once_detail()
    all_categories_eval = _load_all_categories_eval_detail()

    # 最新株価参照（購入株数算定用）
    kabuka_path = _find_latest_kabuka_file()
    kabuka_df = _read_kabuka_wide(kabuka_path)

    # グラフ用：時系列（wide）
    ts_wide = _read_kabuka_ts_wide_for_codes_detail(codes4)

    # --------------------------------------------
    # 現状（owner）のDFを組む：補*は owner=="T" のときのみ
    # --------------------------------------------
    df_owner = target_df.copy()
    df_owner["code"] = df_owner["code"].astype("string").map(lambda v: _norm_code4(v))
    df_owner = df_owner[df_owner["所有者"].astype("string") == owner].copy()

    df_owner["株数"] = pd.to_numeric(df_owner["株数"], errors="coerce")
    df_owner["取得単価"] = pd.to_numeric(df_owner["取得単価"], errors="coerce")
    df_owner["年間配当"] = pd.to_numeric(df_owner["年間配当"], errors="coerce")

    df_owner["投資金額"] = df_owner["株数"] * df_owner["取得単価"]
    df_owner["配当金額"] = df_owner["株数"] * df_owner["年間配当"]

    # 業種ダミー除外
    valid_ind = (
        df_owner["industry_33"].notna()
        & ~df_owner["industry_33"].astype("string").str.strip().isin(["", "-", "0"])
    )
    df_owner = df_owner[valid_ind].copy()

    # 補*列：投資金額_補（owner=="T" かつ 実現益>0 のときだけ控除）
    if owner == "T":
        df_owner["name"] = df_owner.get("name", "")
        df_owner["_pl"] = df_owner.apply(
            lambda r: _get_realized_pl(str(r["code"]), str(r.get("name", "")), pl_by_code, pl_by_name),
            axis=1
        )
        df_owner["_pl_pos"] = np.where(df_owner["_pl"] > 0, df_owner["_pl"], 0.0)
        df_owner["投資金額_補"] = df_owner["投資金額"] - df_owner["_pl_pos"]
    else:
        df_owner["投資金額_補"] = df_owner["投資金額"]  # 互換のため列は持たせる

    # 現状スコア
    base = _calc_portfolio_score(df_owner, all_categories_eval)

    print("\n" + "=" * 100)
    print(f"detail：指定銘柄まとめ（owner={owner}）")
    print(f"株価参照: {os.path.basename(kabuka_path)}")
    print(
        f"現状スコア: {base['スコア_合計(100)']:.2f} / 100"
        f"（配当 {base['スコア_配当利回り(40)']:.2f}/40, 業種 {base['スコア_業種(40)']:.2f}/40, 銘柄 {base['スコア_銘柄(20)']:.2f}/20）"
    )
    print("=" * 100)

    # --------------------------------------------
    # 銘柄ごとの表（1枚）
    # --------------------------------------------
    rows = []
    code_to_name_for_plot: Dict[str, str] = {}

    for code in codes4:
        name = code_to_name.get(code, "")
        ind = code_to_ind.get(code, "")
        divps = float(code_to_divps.get(code, 0.0))  # 1株配当（年）

        code_to_name_for_plot[code] = name

        price = float(_latest_price_from_kabuka(kabuka_df, code))
        buy_shares = int(_calc_buy_shares(price))
        buy_invest = float(buy_shares * price) if (buy_shares > 0 and price > 0) else float("nan")

        realized_pl = float(_get_realized_pl(code, name, pl_by_code, pl_by_name))

        # 年間配当（合計）＝株数×1株配当
        annual_div_total = float(buy_shares * divps) if buy_shares > 0 else 0.0

        # 配当利回り（通常）
        y = np.nan
        if pd.notna(buy_invest) and buy_invest > 0:
            y = (annual_div_total / buy_invest) * 100.0

        # 配当利回り（補*）：owner=="T" のときのみ、実現益（>0）を控除して分母を下げる
        invest_sup = buy_invest
        if owner == "T" and realized_pl > 0 and pd.notna(buy_invest):
            invest_sup = buy_invest - realized_pl

        y_sup = np.nan
        if pd.notna(invest_sup) and invest_sup > 0:
            y_sup = (annual_div_total / invest_sup) * 100.0

        # 追加後スコア（ownerのみ）
        add_row = {
            "code": code,
            "industry_33": ind,
            "株数": float(buy_shares),
            "取得単価": float(price) if (price > 0) else np.nan,
            "年間配当": float(divps),
            "投資金額": float(buy_shares * price) if (buy_shares > 0 and price > 0) else 0.0,
            "配当金額": float(buy_shares * divps) if (buy_shares > 0) else 0.0,
            "投資金額_補": float(buy_shares * price) if (buy_shares > 0 and price > 0) else 0.0,  # 追加分は実現益0
            "所有者": "ADD",
            "name": name,
        }
        df_new = pd.concat([df_owner, pd.DataFrame([add_row])], ignore_index=True)
        after = _calc_portfolio_score(df_new, all_categories_eval)

        rows.append({
            "code": code,
            "name": name,
            "業種": ind,
            "年間配当(1株)": float(divps),
            "実現損益": float(realized_pl),
            "追加株数": int(buy_shares),
            "取得単価": float(price) if (price > 0) else np.nan,
            "投資金額": float(buy_invest),
            "年間配当(合計)": float(annual_div_total),
            "配当利回り": float(y) if pd.notna(y) else np.nan,
            "配当利回り(補*)": float(y_sup) if pd.notna(y_sup) else np.nan,
            "Δ総スコア": after["スコア_合計(100)"] - base["スコア_合計(100)"],
            "Δ配当": after["スコア_配当利回り(40)"] - base["スコア_配当利回り(40)"],
            "Δ業種": after["スコア_業種(40)"] - base["スコア_業種(40)"],
            "Δ銘柄": after["スコア_銘柄(20)"] - base["スコア_銘柄(20)"],
        })

    out = pd.DataFrame(rows)

    show_cols = [
        "code", "name", "業種",
        "年間配当(1株)", "実現損益",
        "追加株数", "取得単価", "投資金額",
        "年間配当(合計)", "配当利回り", "配当利回り(補*)",
        "Δ総スコア", "Δ配当", "Δ業種", "Δ銘柄",
    ]
    show_cols = [c for c in show_cols if c in out.columns]
    out = out[show_cols].copy()

    fmt = {
        "年間配当(1株)": "{:.2f}",
        "実現損益": "{:,.0f}",
        "追加株数": "{:,.0f}",
        "取得単価": "{:.2f}",
        "投資金額": "{:,.0f}",
        "年間配当(合計)": "{:,.0f}",
        "配当利回り": "{:.2f}",
        "配当利回り(補*)": "{:.2f}",
        "Δ総スコア": "{:+.2f}",
        "Δ配当": "{:+.2f}",
        "Δ業種": "{:+.2f}",
        "Δ銘柄": "{:+.2f}",
    }
    try:
        display(out.style.format(fmt))
    except Exception:
        display(out)

    # --------------------------------------------
    # グラフ（1枚）：指数化＆凡例下
    # --------------------------------------------
    _plot_price_multi_detail(
        ts_wide=ts_wide,
        codes4=codes4,
        code_to_name=code_to_name_for_plot,
        title=f"株価（指数：{pd.Timestamp.today().year-2}-01-01=1.0） / owner={owner} / codes={len(codes4)}",
    )
```

### (0)共通の関数などの定義

```python
# ============================================================
# #(0) 共通セル：共通関数（正規化/配当pick/補*/評価業種/スコア計算/kabuka）
#   - 方針A：以後、セル2〜4はここを呼ぶだけにする
# ============================================================

import os
import re
import math
import glob
import numpy as np
import pandas as pd

# -----------------------------
# #(0-1) 基本ユーティリティ
# -----------------------------
EXCLUDED_INDUSTRIES = {"日経ETF", "米国ETF"}  # 評価除外（棒グラフ/業種スコア対象外）

def normalize_code_str_series(s: pd.Series) -> pd.Series:
    """
    - 全角→半角
    - strip
    - 4桁ゼロ埋め
    """
    out = (
        s.astype("string")
         .str.strip()
         .str.translate(str.maketrans("０１２３４５６７８９", "0123456789"))
    )
    # 数字のみのものは4桁ゼロ埋め（"1"→"0001"）
    out = out.map(lambda x: str(x).zfill(4) if (x is not None and re.fullmatch(r"\d+", str(x))) else x)
    return out

def to_num_series(x: pd.Series) -> pd.Series:
    s = x.astype("string").str.replace(",", "", regex=False).str.strip()
    return pd.to_numeric(s, errors="coerce")

def to_num_scalar(x) -> float:
    """'1,234' や ' 123 ' のような表記を数値化（失敗時は NaN）"""
    if pd.isna(x):
        return float("nan")
    s = str(x).replace(",", "").strip()
    m = re.search(r"-?\d+(?:\.\d+)?", s)
    if not m:
        return float("nan")
    try:
        return float(m.group(0))
    except Exception:
        return float("nan")

def yen(x: float) -> str:
    if pd.isna(x):
        return "0円"
    return f"{int(round(float(x))):,}円"

def fix_bom_code_column(df: pd.DataFrame) -> pd.DataFrame:
    """列名にBOMが混入した code を code に寄せる"""
    if "code" in df.columns:
        return df
    for c in df.columns:
        if str(c).replace("\ufeff", "") == "code":
            return df.rename(columns={c: "code"})
    return df

# -----------------------------
# #(0-2) IDmap：評価業種リスト
# -----------------------------
def load_idmap(path: str = "Data/01_IDmap.csv") -> pd.DataFrame:
    df = pd.read_csv(path, dtype={"code": "string"}, encoding="utf-8-sig")
    df = fix_bom_code_column(df)
    if "code" not in df.columns:
        raise KeyError("01_IDmap.csv に code 列が見つかりません。")
    df["code"] = normalize_code_str_series(df["code"])
    return df

def load_all_categories_eval(idmap_df: pd.DataFrame, excluded: set[str] = EXCLUDED_INDUSTRIES) -> list[str]:
    if "industry_33" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に industry_33 列が見つかりません。")

    all_categories = (
        idmap_df["industry_33"]
        .astype("string")
        .map(lambda v: v.strip() if isinstance(v, str) else v)
        .dropna()
    )
    all_categories = sorted({v for v in all_categories.unique().tolist() if v not in ["", "-", "0"]})
    all_categories_eval = [c for c in all_categories if c not in excluded]

    if len(all_categories_eval) == 0:
        raise ValueError("評価対象の industry_33 が 0 件になりました。除外設定やIDmapを確認してください。")

    return all_categories_eval

# -----------------------------
# #(0-3) 03_haitou：年間配当 pick（予優先）
# -----------------------------
def build_haitou_pick_df(path: str = "Data/03_haitou.csv") -> pd.DataFrame:
    haitou_df = pd.read_csv(path, dtype={"code": "string"}, encoding="utf-8-sig")
    haitou_df = fix_bom_code_column(haitou_df)
    if "code" not in haitou_df.columns:
        raise KeyError("03_haitou.csv に code 列が見つかりません。")

    haitou_df["code"] = normalize_code_str_series(haitou_df["code"])

    year_cols = [c for c in haitou_df.columns if re.fullmatch(r"\d{4}", str(c))]
    year_cols_sorted = sorted(year_cols, key=lambda x: int(x))

    def _parse_div_cell(x):
        if pd.isna(x):
            return (False, None)
        s = str(x).strip()
        if s == "":
            return (False, None)
        is_forecast = "予" in s
        m = re.search(r"(\d+(?:\.\d+)?)", s)
        if not m:
            return (is_forecast, None)
        return (is_forecast, float(m.group(1)))

    def _pick_latest_dividend(row):
        latest_forecast = (None, None)  # (year, value)
        latest_actual = (None, None)

        for y in year_cols_sorted:
            is_fc, val = _parse_div_cell(row.get(y))
            if val is None:
                continue
            yi = int(y)
            if is_fc:
                if latest_forecast[0] is None or yi > latest_forecast[0]:
                    latest_forecast = (yi, val)
            else:
                if latest_actual[0] is None or yi > latest_actual[0]:
                    latest_actual = (yi, val)

        if latest_forecast[0] is not None:
            return pd.Series({"年間配当": latest_forecast[1], "年間配当_年度": latest_forecast[0], "年間配当_区分": "予想"})
        if latest_actual[0] is not None:
            return pd.Series({"年間配当": latest_actual[1], "年間配当_年度": latest_actual[0], "年間配当_区分": "実績"})
        return pd.Series({"年間配当": pd.NA, "年間配当_年度": pd.NA, "年間配当_区分": pd.NA})

    picked = haitou_df.apply(_pick_latest_dividend, axis=1)
    haitou_pick_df = pd.concat([haitou_df[["code"]].copy(), picked], axis=1)
    haitou_pick_df = (
        haitou_pick_df
        .sort_values(["code", "年間配当_年度"])
        .drop_duplicates(subset=["code"], keep="last")
    )
    return haitou_pick_df

# -----------------------------
# #(0-4) summary.csv：実現損益マップ（2行目）
# -----------------------------
def load_realized_pl_maps(path: str = "Data/jyouto/summary.csv") -> tuple[dict[str, float], dict[str, float]]:
    """
    2行目(index=1)を読む
      - key=4桁コード（正規化済み） -> 実現損益
      - key=銘柄名 -> 実現損益（補助）
    """
    if not os.path.exists(path):
        return {}, {}

    sdf = pd.read_csv(path, dtype="string", encoding="utf-8-sig")
    if len(sdf) < 2:
        return {}, {}

    row2 = sdf.iloc[1]
    pl_by_code: dict[str, float] = {}
    pl_by_name: dict[str, float] = {}

    for k, v in row2.items():
        if k is None:
            continue
        key_raw = str(k).strip()
        if key_raw == "":
            continue

        val_num = to_num_scalar(v)
        if pd.isna(val_num):
            continue

        key_norm = str(pd.Series([key_raw], dtype="string").pipe(normalize_code_str_series).iloc[0])
        if re.fullmatch(r"\d{4}", key_norm):
            pl_by_code[key_norm] = float(val_num)
        else:
            pl_by_name[key_raw] = float(val_num)

    return pl_by_code, pl_by_name

# -----------------------------
# #(0-5) 補*：投資金額_補（Tのみ・益のみ控除）
# -----------------------------
def apply_investment_supplement(
    df: pd.DataFrame,
    owner_col: str = "所有者",
    code_col: str = "code",
    name_col: str = "name",
    invest_col: str = "投資金額",
    sup_col: str = "投資金額_補",
    realized_col: str = "実現損益",
    summary_path: str = "Data/jyouto/summary.csv",
) -> pd.DataFrame:
    """
    - df に「実現損益」「実現益_反映」「投資金額_補」を付与
    - 実現益を反映するのは owner=='T' かつ realized>0 のときのみ
    - 損失（<=0）は反映しない
    - 投資金額_補 は最低0にクリップ
    """
    out = df.copy()
    if invest_col not in out.columns:
        raise KeyError(f"apply_investment_supplement: '{invest_col}' 列が必要です。")
    if owner_col not in out.columns:
        raise KeyError(f"apply_investment_supplement: '{owner_col}' 列が必要です。")
    if code_col not in out.columns:
        raise KeyError(f"apply_investment_supplement: '{code_col}' 列が必要です。")

    pl_by_code, pl_by_name = load_realized_pl_maps(summary_path)

    out[code_col] = normalize_code_str_series(out[code_col])

    realized = out[code_col].astype("string").map(pl_by_code)
    if name_col in out.columns and len(pl_by_name) > 0:
        realized = realized.fillna(out[name_col].map(pl_by_name))

    out[realized_col] = pd.to_numeric(realized, errors="coerce").fillna(0.0)

    out["実現益_反映"] = np.where(
        (out[owner_col].astype("string") == "T") & (out[realized_col] > 0),
        out[realized_col],
        0.0
    )

    inv = pd.to_numeric(out[invest_col], errors="coerce").fillna(0.0)
    ref = pd.to_numeric(out["実現益_反映"], errors="coerce").fillna(0.0)
    out[sup_col] = (inv - ref).clip(lower=0.0)

    return out

# -----------------------------
# #(0-6) スコア計算（入口は calc_portfolio_score のみ）
#   - 配当：補*利回り（投資金額_補）を採点に使用（Tのみ・益のみ控除）
#   - 銘柄集中：補*投資金額比率（投資金額_補）を使用
#   - 業種：配当金額ベース（ETF除外込み）を使用
# -----------------------------
def _score_dividend_yield_linear(y_pct: float, y0: float = 4.0, y1: float = 10.0, max_score: float = 40.0) -> float:
    if y_pct <= y0:
        return 0.0
    if y_pct >= y1:
        return float(max_score)
    return float((y_pct - y0) / (y1 - y0) * max_score)

def _score_industries_piecewise_by_dividend_share(
    work_df: pd.DataFrame,
    all_categories_eval: list[str],
    industry_col: str = "industry_33",
    dividend_amount_col: str = "配当金額",
) -> float:
    total_dividend = float(pd.to_numeric(work_df[dividend_amount_col], errors="coerce").fillna(0.0).sum())
    if total_dividend <= 0:
        return 0.0

    share = (
        work_df.groupby(industry_col)[dividend_amount_col].sum()
        / total_dividend
        * 100.0
    )
    share = share.reindex(all_categories_eval, fill_value=0.0)

    n = len(all_categories_eval)
    per_max = 40.0 / n

    def piecewise_points(r_pct: float) -> float:
        r = float(r_pct)
        if r <= 0.0:
            return 0.0
        if 0.0 < r < 1.5:
            return per_max * (r / 1.5)
        if 1.5 <= r <= 15.0:
            return per_max
        if 15.0 < r < 17.0:
            return per_max * (1.0 - (r - 15.0) / 2.0)
        return 0.0

    score = sum(piecewise_points(r) for r in share.values)
    return float(max(0.0, min(40.0, score)))

def _score_stocks_by_invest_concentration(
    work_df: pd.DataFrame,
    invest_col: str = "投資金額_補",
    code_col: str = "code",
    cap: float = 0.03,
) -> float:
    total = float(pd.to_numeric(work_df[invest_col], errors="coerce").fillna(0.0).sum())
    if total <= 0:
        return 0.0

    by_code = work_df.groupby(code_col)[invest_col].sum()
    weights = by_code / total

    excess = (weights - cap).clip(lower=0.0)
    penalty_points = float((excess * 100.0).sum())

    score = 20.0 - penalty_points
    return float(max(0.0, score))

def calc_portfolio_score(
    work_df: pd.DataFrame,
    all_categories_eval: list[str],
    *,
    industry_col: str = "industry_33",
    code_col: str = "code",
    invest_col_raw: str = "投資金額",
    invest_col_sup: str = "投資金額_補",
    dividend_amount_col: str = "配当金額",
    use_sup_invest: bool = True,
    yield_y0: float = 4.0,      # 4% → 0点
    yield_y1: float = 10.0,     # 10% → 40点
) -> dict:
    w = work_df.copy()

    for c in [invest_col_raw, dividend_amount_col]:
        if c not in w.columns:
            raise KeyError(f"calc_portfolio_score: '{c}' 列が必要です。")

    w[invest_col_raw] = pd.to_numeric(w[invest_col_raw], errors="coerce").fillna(0.0)
    w[dividend_amount_col] = pd.to_numeric(w[dividend_amount_col], errors="coerce").fillna(0.0)

    if use_sup_invest:
        if invest_col_sup not in w.columns:
            raise KeyError(f"calc_portfolio_score(use_sup_invest=True): '{invest_col_sup}' がありません。apply_investment_supplement を先に適用してください。")
        w[invest_col_sup] = pd.to_numeric(w[invest_col_sup], errors="coerce").fillna(0.0)
        invest_used = float(w[invest_col_sup].sum())
    else:
        invest_used = float(w[invest_col_raw].sum())

    invest_raw = float(w[invest_col_raw].sum())
    total_dividend = float(w[dividend_amount_col].sum())

    y_raw = (total_dividend / invest_raw * 100.0) if invest_raw > 0 else 0.0
    y_sup = (total_dividend / invest_used * 100.0) if invest_used > 0 else 0.0

    score_y = _score_dividend_yield_linear(y_sup, y0=yield_y0, y1=yield_y1, max_score=40.0)

    score_i = _score_industries_piecewise_by_dividend_share(
        w,
        all_categories_eval=all_categories_eval,
        industry_col=industry_col,
        dividend_amount_col=dividend_amount_col,
    )

    invest_for_stock = invest_col_sup if use_sup_invest else invest_col_raw
    score_s = _score_stocks_by_invest_concentration(
        w,
        invest_col=invest_for_stock,
        code_col=code_col,
        cap=0.03,
    )

    return {
        "total_invest_raw": invest_raw,
        "total_invest_used": invest_used,
        "total_dividend": total_dividend,
        "dividend_yield_pct": float(y_raw),
        "dividend_yield_sup_pct": float(y_sup),
        "score_yield": float(score_y),
        "score_industry": float(score_i),
        "score_stock": float(score_s),
        "score_total": float(score_y + score_i + score_s),
    }

# -----------------------------
# #(0-7) kabuka：最新年/終値/時系列wide（detail用）
# -----------------------------
def find_latest_kabuka_file(base_dir: str = os.path.join("Data", "kabuka")) -> str:
    if not os.path.isdir(base_dir):
        raise FileNotFoundError(f"フォルダが見つかりません: {base_dir}")

    years = []
    for fn in os.listdir(base_dir):
        m = re.fullmatch(r"(\d{4})\.csv", fn)
        if m:
            years.append(int(m.group(1)))
    if not years:
        raise FileNotFoundError("Data/kabuka に YYYY.csv（例: 2026.csv）が見つかりません。")

    y = max(years)
    return os.path.join(base_dir, f"{y}.csv")

def read_kabuka_wide(path: str) -> pd.DataFrame:
    """
    kabuka/YYYY.csv は「先頭列=日付(idx)」前提。
    """
    kdf = pd.read_csv(path, encoding="utf-8-sig")

    first_col = kdf.columns[0]
    idx_col = "idx"
    if first_col != idx_col:
        kdf = kdf.rename(columns={first_col: idx_col})

    return kdf

def latest_price_from_kabuka(kdf: pd.DataFrame, code_str: str) -> float:
    col = str(code_str)
    if col not in kdf.columns:
        return float("nan")
    s = pd.to_numeric(kdf[col], errors="coerce").dropna()
    if len(s) == 0:
        return float("nan")
    return float(s.iloc[-1])

def calc_buy_shares(price: float, unit: int = 100, min_yen: float = 200000.0) -> int:
    if price is None or np.isnan(price) or price <= 0:
        return 0
    lot_cost = price * unit
    if lot_cost > min_yen:
        return unit
    k = math.floor(min_yen / lot_cost) + 1
    return int(unit * k)

def read_kabuka_ts_wide_for_codes(
    codes: list[str],
    start_date: pd.Timestamp | None = None,
    end_date: pd.Timestamp | None = None,
    base_dir: str = os.path.join("Data", "kabuka"),
) -> pd.DataFrame:
    if not isinstance(codes, list) or len(codes) == 0:
        return pd.DataFrame(columns=["date"])

    today = pd.Timestamp.today().normalize()
    if start_date is None:
        start_date = pd.Timestamp(year=today.year - 2, month=1, day=1)
    if end_date is None:
        end_date = today

    code_cols = [
        str(pd.Series([c], dtype="string").pipe(normalize_code_str_series).iloc[0])
        for c in codes
    ]

    paths = sorted(glob.glob(os.path.join(base_dir, "[0-9][0-9][0-9][0-9].csv")))
    if len(paths) == 0:
        return pd.DataFrame(columns=["date"] + code_cols)

    chunks = []
    for p in paths:
        kdf = pd.read_csv(p, encoding="utf-8-sig")

        first_col = kdf.columns[0]
        idx_col = "idx"
        if first_col != idx_col:
            kdf = kdf.rename(columns={first_col: idx_col})

        present = [c for c in code_cols if c in kdf.columns]
        if len(present) == 0:
            continue

        mask = kdf[idx_col].astype("string").str.match(r"^\d{4}-\d{2}-\d{2}$", na=False)
        tmp = kdf.loc[mask, [idx_col] + present].copy()

        tmp[idx_col] = pd.to_datetime(tmp[idx_col], errors="coerce")
        tmp = tmp.dropna(subset=[idx_col])
        if len(tmp) == 0:
            continue

        tmp = tmp[(tmp[idx_col] >= start_date) & (tmp[idx_col] <= end_date)]
        if len(tmp) == 0:
            continue

        for c in present:
            tmp[c] = pd.to_numeric(tmp[c], errors="coerce")

        tmp = tmp.rename(columns={idx_col: "date"})
        chunks.append(tmp)

    if len(chunks) == 0:
        return pd.DataFrame(columns=["date"] + code_cols)

    out = pd.concat(chunks, ignore_index=True)
    out = out.sort_values("date").drop_duplicates(subset=["date"], keep="last").reset_index(drop=True)

    for c in code_cols:
        if c not in out.columns:
            out[c] = np.nan

    out = out[["date"] + code_cols]
    return out
```

### (1)投資金額サマリー

```python
# ============================================================
# 【このセル単体で完結】保有銘柄分析（全体 / 所有者別 / 所有者×証券会社×口座別）
# - #(0) 共通セルを前提（正規化/配当pick/補* 等は再定義しない）
# - 02_OWNlist × 01_IDmap × 03_haitou を統合
# - 集計は「全銘柄」を対象（配当欠損は 0 扱いで年間配当金額を計算）
# - 出力：各ブロックごとに
#   ①投資金額・年間配当（テキスト2行）
#   ②銘柄一覧（code/所有者/証券会社/口座区分/株数/取得単価/name/industry_33/年間配当/年度/区分/実現損益）
#   ③業種割合（投資金額ベース）円グラフ
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

# ============================================================
# #(1) 表示設定（省略「...」を出さない）
# ============================================================
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 0)
pd.set_option("display.max_colwidth", None)

# ============================================================
# #(2) 読み込み：02_OWNlist × 01_IDmap（証券会社列も自動判定）
#   ※ BOM対策や code 正規化は #(0) の共通関数を使用
# ============================================================
own_df = pd.read_csv("Data/02_OWNlist.csv", dtype={"code": "string"}, encoding="utf-8-sig")
idmap_df = load_idmap("Data/01_IDmap.csv")  # #(0) 共通：BOM対策込み

own_df = fix_bom_code_column(own_df)
own_df["code"] = normalize_code_str_series(own_df["code"])
idmap_df["code"] = normalize_code_str_series(idmap_df["code"])

# --- 証券会社列の自動判定（broker優先、無ければ「証券会社」） ---
if "broker" in own_df.columns:
    broker_col = "broker"
elif "証券会社" in own_df.columns:
    broker_col = "証券会社"
else:
    raise KeyError("02_OWNlist.csv に 'broker' または '証券会社' 列が見つかりません。列名を確認してください。")

# 必須列チェック
need_own = ["code", "所有者", broker_col, "口座区分", "株数", "取得単価"]
need_idm = ["code", "name", "industry_33"]
for c in need_own:
    if c not in own_df.columns:
        raise KeyError(f"02_OWNlist.csv に必要列 '{c}' がありません。")
for c in need_idm:
    if c not in idmap_df.columns:
        raise KeyError(f"01_IDmap.csv に必要列 '{c}' がありません。")

own_df = own_df[need_own].rename(columns={broker_col: "証券会社"})
idmap_df = idmap_df[need_idm]

base_df = pd.merge(own_df, idmap_df, on="code", how="left")
base_df["industry_33"] = base_df["industry_33"].fillna("不明")  # 円グラフ用

# ============================================================
# #(3) 03_haitou：年間配当 pick を作成して base_df に付与
#   ※ #(0) 共通の build_haitou_pick_df を使用（再定義しない）
# ============================================================
haitou_pick_df = build_haitou_pick_df("Data/03_haitou.csv")
analyze_df = pd.merge(base_df, haitou_pick_df, on="code", how="left")

# ============================================================
# #(4) 投資金額・年間配当金額・補*（投資金額_補）・配当利回り（計算のみ）
#   - 配当欠損は 0 扱いで年間配当金額を計算
#   - 補* は owner='T' かつ 実現益>0 のときのみ控除（#(0) の apply_investment_supplement）
# ============================================================
analyze_df["株数_num"] = to_num_series(analyze_df["株数"])
analyze_df["取得単価_num"] = to_num_series(analyze_df["取得単価"])
analyze_df["年間配当_num"] = pd.to_numeric(analyze_df["年間配当"], errors="coerce").fillna(0.0)

analyze_df["投資金額"] = analyze_df["株数_num"] * analyze_df["取得単価_num"]
analyze_df["年間配当金額"] = analyze_df["株数_num"] * analyze_df["年間配当_num"]

# 補*（実現損益/投資金額_補）を一括生成（Tのみ益のみ控除、投資金額_補は0下限）
analyze_df = apply_investment_supplement(
    analyze_df,
    owner_col="所有者",
    code_col="code",
    name_col="name",
    invest_col="投資金額",
    sup_col="投資金額_補",
    realized_col="総合実現損益",   # このセルでは列名を「総合実現損益」に合わせておく
    summary_path="Data/jyouto/summary.csv",
)

# 利回り（計算のみ。表示はしない）
analyze_df["配当利回り"] = np.where(
    (pd.to_numeric(analyze_df["投資金額_補"], errors="coerce").fillna(0.0) > 0),
    analyze_df["年間配当金額"] / analyze_df["投資金額_補"],
    np.nan
)

# ============================================================
# #(5) 表示：テキスト2行（投資金額・年間配当）→ 一覧表 → 円グラフ
#   - 配当利回りは「計算するが表示しない」
#   - 一覧表の列：
#     code, name, 所有者, 証券会社, 口座区分, 株数, 取得単価, 配当, 配当年度, 配当区分, 業種, 実現損益
#   - 実現損益は所有者=Tのときだけ表示。それ以外は 0
#   - 円グラフは「投資金額が計算できる行のみ」で投資金額ベース業種割合
# ============================================================
DISPLAY_COLS = [
    "code", "name", "所有者", "証券会社", "口座区分", "株数", "取得単価",
    "配当", "配当年度", "配当区分", "業種", "実現損益"
]

def _yen2(x) -> str:
    if x is None or pd.isna(x):
        return "NaN"
    return f"{int(round(float(x))):,}円"

def show_portfolio_block(df: pd.DataFrame, title: str):
    print("\n" + "=" * 100)
    print(title)
    print("=" * 100)

    # 集計・円グラフ用：投資金額が計算できるものだけ
    df_valid_inv = df[df["投資金額"].notna() & (df["投資金額"] > 0)].copy()

    total_inv = float(df_valid_inv["投資金額"].sum())
    total_div = float(df_valid_inv["年間配当金額"].sum())

    # テキストは2行（利回りは出さない）
    print(f"投資金額: {_yen2(total_inv)}")
    print(f"年間配当: {_yen2(total_div)}")

    # ---- 一覧表 ----
    disp = df.copy()

    disp["配当"] = disp["年間配当"] if "年間配当" in disp.columns else pd.NA
    disp["配当年度"] = disp["年間配当_年度"] if "年間配当_年度" in disp.columns else pd.NA
    disp["配当区分"] = disp["年間配当_区分"] if "年間配当_区分" in disp.columns else pd.NA
    disp["業種"] = disp["industry_33"] if "industry_33" in disp.columns else pd.NA

    # 実現損益：所有者=T のときだけ（それ以外は0）
    if "総合実現損益" in disp.columns:
        disp["実現損益"] = np.where(
            disp["所有者"].astype("string") == "T",
            pd.to_numeric(disp["総合実現損益"], errors="coerce").fillna(0.0),
            0.0
        )
    else:
        disp["実現損益"] = 0.0

    cols = [c for c in DISPLAY_COLS if c in disp.columns]
    display(disp[cols])

    # ---- 円グラフ（投資金額ベース業種割合）----
    pie_src = (
        df_valid_inv.groupby("industry_33", dropna=False)["投資金額"]
        .sum()
        .sort_values(ascending=False)
    )

    if len(pie_src) == 0:
        print("円グラフ: 投資金額が計算できる銘柄がありません（株数/取得単価の欠損・形式を確認してください）。")
        return

    plt.figure()
    plt.title(f"{title}\n（投資金額ベース業種割合 / 合計: {_yen2(total_inv)}）")
    plt.pie(
        pie_src.values,
        labels=pie_src.index,
        autopct="%1.1f%%",
        startangle=90,
        counterclock=False
    )
    plt.axis("equal")
    plt.show()

# ============================================================
# #(6) (1)全体 → (2)所有者別
# ============================================================
show_portfolio_block(analyze_df, "【(1) 全体】")

for owner, df_owner in analyze_df.groupby("所有者", dropna=False):
    show_portfolio_block(df_owner, f"【(2) 所有者別】所有者={owner}")
```

### (2)配当銘柄分析

```python
# ============================================================
# 【このセル単体で完結】保有銘柄 → 条件抽出(target_df) → 配当分析（円グラフ2列＋横棒＋スコア）
#
# 重要：
#   - #(0) 共通セルを前提（正規化/配当pick/補*/評価業種/スコア計算は再定義しない）
#   - スコアは calc_portfolio_score のみ（入口は1つ）
#   - 補*は apply_investment_supplement のみ（Tのみ・益のみ控除）
#
# 入力CSV:
#   Data/02_OWNlist.csv  : code, 所有者, (broker or 証券会社), 口座区分, 株数, 取得単価
#   Data/01_IDmap.csv    : code, name, industry_33
#   Data/03_haitou.csv   : code, 2019, 2020, ... （セルに「予」あり/なし）
#
# 抽出条件（targets）:
#   - 年間配当があり、かつ 年間配当_年度 が「今年 or 昨年」
#   - (所有者, 証券会社, 口座区分) の組が targets に一致（None または "*" はワイルドカード）
#
# 出力:
#   - 全体：円グラフ（投資金額(補*)/配当金額）＋横棒（配当金額ベース%）＋スコア
#   - 所有者別：同様
# ============================================================

import pandas as pd
import numpy as np
import datetime as _dt
import matplotlib.pyplot as plt
import japanize_matplotlib

# ============================================================
# #(1) 表示設定（省略「...」を出さない）
# ============================================================
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 0)
pd.set_option("display.max_colwidth", None)

# ============================================================
# #(2) CSV読み込み & 統合（02_OWNlist × 01_IDmap × 03_haitou_pick）
#   - #(0) の load_idmap / build_haitou_pick_df / normalize_code_str_series を使用
# ============================================================
own_df = pd.read_csv("Data/02_OWNlist.csv", dtype={"code": "string"}, encoding="utf-8-sig")
own_df = fix_bom_code_column(own_df)
own_df["code"] = normalize_code_str_series(own_df["code"])

idmap_df = load_idmap("Data/01_IDmap.csv")  # BOM対策込み
idmap_df["code"] = normalize_code_str_series(idmap_df["code"])

# --- 証券会社列の自動判定（broker優先、無ければ「証券会社」） ---
if "broker" in own_df.columns:
    broker_col = "broker"
elif "証券会社" in own_df.columns:
    broker_col = "証券会社"
else:
    raise KeyError("02_OWNlist.csv に 'broker' または '証券会社' 列が見つかりません。列名を確認してください。")

need_own = ["code", "所有者", broker_col, "口座区分", "株数", "取得単価"]
need_idm = ["code", "name", "industry_33"]
for c in need_own:
    if c not in own_df.columns:
        raise KeyError(f"02_OWNlist.csv に必要列 '{c}' がありません。")
for c in need_idm:
    if c not in idmap_df.columns:
        raise KeyError(f"01_IDmap.csv に必要列 '{c}' がありません。")

own_df = own_df[need_own].rename(columns={broker_col: "証券会社"})
idmap_df = idmap_df[need_idm]

merged_df = pd.merge(own_df, idmap_df, on="code", how="left")

# 年間配当 pick（予優先）
haitou_pick_df = build_haitou_pick_df("Data/03_haitou.csv")
merged_df = pd.merge(merged_df, haitou_pick_df, on="code", how="left")

# ============================================================
# #(3) 抽出関数（targets）
# ============================================================
def filter_target_stocks(
    merged_df: pd.DataFrame,
    owners_brokers_accounts: list[tuple],
) -> pd.DataFrame:

    CURRENT_YEAR = _dt.date.today().year
    TARGET_YEARS = {CURRENT_YEAR, CURRENT_YEAR - 1}

    base_df = merged_df[
        merged_df["年間配当"].notna()
        & merged_df["年間配当_年度"].notna()
        & merged_df["年間配当_年度"].astype("Int64").isin(TARGET_YEARS)
    ].copy()

    cond = False
    for owner, broker, account in owners_brokers_accounts:
        c = True
        if owner not in (None, "*"):
            c = c & (base_df["所有者"] == owner)
        if broker not in (None, "*"):
            c = c & (base_df["証券会社"] == broker)
        if account not in (None, "*"):
            c = c & (base_df["口座区分"] == account)
        cond = cond | c

    return base_df[cond].copy()

# 対象口座の指定（必要に応じて編集）
targets = [
    ("Y", "SBI", "NISA"),
    ("Y", "SBI", "特定"),
    ("T", "SBI", "NISA"),
    ("T", "SBI", "特定"),
]

target_df = filter_target_stocks(merged_df, targets)

print("=== 条件一致・集計対象銘柄 ===")
print("件数:", len(target_df))
display(target_df)

# ============================================================
# #(4) 事前整形：金額列作成 + 補*（投資金額_補）を付与
#   - 補*は apply_investment_supplement に一本化（Tのみ・益のみ控除）
#   - スコアの配当利回りは「配当金額 ÷ 投資金額_補」を採点に使用（calc_portfolio_score）
# ============================================================
df = target_df.copy()

# 数値化
df["株数"] = to_num_series(df["株数"])
df["取得単価"] = to_num_series(df["取得単価"])
df["年間配当"] = pd.to_numeric(df["年間配当"], errors="coerce")

# 金額列
df["投資金額"] = df["株数"] * df["取得単価"]
df["配当金額"] = df["株数"] * df["年間配当"]

# industry_33 の無効値を落とす（円グラフ/スコアの分母を汚さない）
valid_industry = (
    df["industry_33"].notna()
    & ~df["industry_33"].astype("string").str.strip().isin(["", "-", "0"])
)
df = df[valid_industry].copy()

# 補*付与（総合実現損益、実現益_反映、投資金額_補）
df = apply_investment_supplement(
    df,
    owner_col="所有者",
    code_col="code",
    name_col="name",
    invest_col="投資金額",
    sup_col="投資金額_補",
    realized_col="総合実現損益",
    summary_path="Data/jyouto/summary.csv",
)

# 表示用：実現損益は T のときだけ（それ以外は 0）
df["実現損益"] = np.where(
    df["所有者"].astype("string") == "T",
    pd.to_numeric(df["総合実現損益"], errors="coerce").fillna(0.0),
    0.0
)

# 表出力（列順）
df["配当"] = df["年間配当"]
df["配当年度"] = df["年間配当_年度"]
df["業種"] = df["industry_33"]

DISPLAY_COLS = [
    "code", "name", "所有者", "証券会社", "口座区分",
    "株数", "取得単価",
    "配当", "配当年度", "業種",
    "実現損益",
    "投資金額", "投資金額_補", "配当金額",
]
print("=== 条件一致・集計対象銘柄（表示用） ===")
print("件数:", len(df))
display(df[[c for c in DISPLAY_COLS if c in df.columns]])

# ============================================================
# #(5) 評価業種リスト（棒グラフの軸） ※ #(0) の関数で一元化
#   - EXCLUDED_INDUSTRIES（{日経ETF, 米国ETF}）は除外済み
# ============================================================
all_categories_eval = load_all_categories_eval(idmap_df, excluded=EXCLUDED_INDUSTRIES)

# プロットでは「除外業種」は表示対象から外す（見た目とスコアの整合）
df_plot = df[~df["industry_33"].astype("string").isin(EXCLUDED_INDUSTRIES)].copy()

# ============================================================
# #(6) 円グラフ（2列）＋スコア表示
#   - 左：投資金額ベース（補*） -> 投資金額_補
#   - 右：配当金額ベース -> 配当金額
#   - 右側にスコア内訳（calc_portfolio_score）を表示
#   - 円グラフ：降順 / 12時開始 / 時計回り
# ============================================================
def plot_two_pies(work_df: pd.DataFrame, title_prefix: str):
    fig, axes = plt.subplots(1, 2, figsize=(14, 7))

    info = calc_portfolio_score(
        work_df,
        all_categories_eval=all_categories_eval,
        industry_col="industry_33",
        code_col="code",
        invest_col_raw="投資金額",
        invest_col_sup="投資金額_補",
        dividend_amount_col="配当金額",
        use_sup_invest=True,
        yield_y0=4.0,
        yield_y1=10.0,
    )

    for ax, value_col, subtitle in zip(
        axes,
        ["投資金額_補", "配当金額"],
        ["投資金額ベース（補*）", "配当金額ベース"],
    ):
        by_ind = work_df.groupby("industry_33")[value_col].sum()
        by_ind = by_ind[by_ind > 0].sort_values(ascending=False)

        total = float(by_ind.sum())

        if len(by_ind) == 0:
            ax.text(0.5, 0.5, "データなし", ha="center", va="center")
            ax.set_title(f"{subtitle}\n（合計 {yen(0)}）")
            ax.axis("off")
            continue

        ax.pie(
            by_ind.values,
            labels=by_ind.index,
            autopct="%1.1f%%",
            startangle=90,
            counterclock=False,
        )
        ax.axis("equal")

        if value_col == "配当金額":
            ax.set_title(
                f"{subtitle}\n"
                f"（合計 {yen(total)} / 配当利回り {info['dividend_yield_pct']:.2f}%"
                f" / 配当利回り(補*) {info['dividend_yield_sup_pct']:.2f}%）"
            )
            text = (
                f"スコア {info['score_total']:.1f} / 100\n"
                f"配当利回り {info['score_yield']:.1f} / 40\n"
                f"業種 {info['score_industry']:.1f} / 40\n"
                f"銘柄 {info['score_stock']:.1f} / 20"
            )
            ax.text(0, 0, text, ha="center", va="center", fontsize=11)
        else:
            ax.set_title(f"{subtitle}\n（合計 {yen(total)}）")

    fig.suptitle(title_prefix, fontsize=14)
    plt.tight_layout()
    plt.show()

# ============================================================
# #(7) 横棒グラフ：業種割合（%）
#   - 配当金額ベース（※補*は分母＝投資金額側にのみ効かせる設計）
#   - 軸は all_categories_eval（該当なしは0%）
# ============================================================
def plot_dividend_barh_pct(work_df: pd.DataFrame, title_prefix: str):
    s = work_df.groupby("industry_33")["配当金額"].sum()
    total = float(s.sum())
    if total <= 0:
        print(f"{title_prefix}：配当金額が0のため棒グラフを表示できません。")
        return

    pct = (s / total) * 100.0
    pct = pct.reindex(all_categories_eval, fill_value=0.0)
    pct = pct.sort_values(ascending=True)  # barhは昇順で下が最大

    fig_height = max(8, len(pct) * 0.35)
    plt.figure(figsize=(9, fig_height))
    plt.barh(pct.index, pct.values)
    plt.title(f"{title_prefix}：配当金額ベース 業種割合（%）")
    plt.xlabel("割合（%）")
    plt.xlim(0, max(1.0, float(pct.max()) * 1.1))
    plt.tight_layout()
    plt.show()

# ============================================================
# #(8) 実行：全体 → 所有者別
# ============================================================
plot_two_pies(df_plot, "全体：業種割合（円グラフ）")
plot_dividend_barh_pct(df_plot, "全体")

owners = sorted(df_plot["所有者"].astype("string").dropna().unique().tolist())
for owner in owners:
    dfo = df_plot[df_plot["所有者"] == owner].copy()
    plot_two_pies(dfo, f"所有者 {owner}：業種割合（円グラフ）")
    plot_dividend_barh_pct(dfo, f"所有者 {owner}")
```

```python
# ============================================================
# 【4つ目セル】銘柄詳細：detail(["1301","1305"], "T") のように複数コードを一括表示（共通セル方針A）
#
# 前提（#(0) 共通セルの存在）：
#   - target_df が存在（2つ目セルで作成済み）
#   - 以下の関数/定数が # (0) に存在：
#       normalize_code_str_series
#       fix_bom_code_column
#       load_idmap
#       build_haitou_pick_df
#       load_all_categories_eval
#       find_latest_kabuka_file, read_kabuka_wide, latest_price_from_kabuka
#       calc_buy_shares
#       apply_investment_supplement
#       calc_portfolio_score
#       EXCLUDED_INDUSTRIES（無ければこのセルでfallback）
#
# 補*（共通セル定義）：
#   - 所有者=T かつ 実現損益>0 のときだけ投資金額を減算（損失は無視）
# ============================================================

# ============================================================
# #(1) imports
# ============================================================
import os
import re
import glob
from typing import List, Dict, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# ============================================================
# #(2) fallback（EXCLUDED_INDUSTRIES が無い場合だけ）
# ============================================================
if "EXCLUDED_INDUSTRIES" not in globals():
    EXCLUDED_INDUSTRIES = {"日経ETF", "米国ETF"}  # fallback

# ============================================================
# #(3) 依存チェック（足りないときはここで分かりやすく止める）
# ============================================================
_required_globals = [
    "target_df",
    "normalize_code_str_series",
    "fix_bom_code_column",
    "load_idmap",
    "build_haitou_pick_df",
    "load_all_categories_eval",
    "find_latest_kabuka_file",
    "read_kabuka_wide",
    "latest_price_from_kabuka",
    "calc_buy_shares",
    "apply_investment_supplement",
    "calc_portfolio_score",
]
_missing = [name for name in _required_globals if name not in globals()]
if _missing:
    raise NameError(
        "4つ目セルの前提が不足しています。上から順に実行できているか確認してください。不足: "
        + ", ".join(_missing)
    )

# ============================================================
# #(4) このセル内専用：コード正規化（detail入力の揺れ吸収用）
#   - 全角→半角
#   - strip
#   - 数字のみで長さ<=4 は左ゼロ埋めで4桁化（kabuka列/IDmap/haitouに合わせやすく）
#   - 5桁以上（ETF等）はそのまま
# ============================================================
def _norm_code4(x: str) -> str:
    s = str(x).strip().translate(str.maketrans("０１２３４５６７８９", "0123456789"))
    if re.fullmatch(r"\d+", s) and len(s) <= 4:
        return s.zfill(4)
    return s

def _norm_codes4(codes: List[str]) -> List[str]:
    return [_norm_code4(c) for c in codes]

# ============================================================
# #(5) マスタ辞書（code -> name/industry/divps）を1回だけ作る
#   - IDmap + haitou_pick（予優先）を結合
# ============================================================
def _build_master_maps_detail() -> tuple[Dict[str, str], Dict[str, str], Dict[str, float]]:
    idmap_df = load_idmap(os.path.join("Data", "01_IDmap.csv"))
    idmap_df["code"] = normalize_code_str_series(idmap_df["code"]).map(_norm_code4)

    if "name" not in idmap_df.columns:
        idmap_df["name"] = ""
    if "industry_33" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に industry_33 列が見つかりません。")

    haitou_pick_df = build_haitou_pick_df(os.path.join("Data", "03_haitou.csv"))
    haitou_pick_df["code"] = normalize_code_str_series(haitou_pick_df["code"]).map(_norm_code4)
    haitou_pick_df["年間配当"] = pd.to_numeric(haitou_pick_df["年間配当"], errors="coerce")

    m = pd.merge(
        idmap_df[["code", "name", "industry_33"]],
        haitou_pick_df[["code", "年間配当"]],
        on="code",
        how="left",
    )
    m["年間配当"] = pd.to_numeric(m["年間配当"], errors="coerce").fillna(0.0)

    code_to_name = dict(zip(m["code"].astype(str), m["name"].fillna("").astype(str)))
    code_to_ind = dict(zip(m["code"].astype(str), m["industry_33"].fillna("").astype(str)))
    code_to_divps = dict(zip(m["code"].astype(str), m["年間配当"].astype(float)))  # 1株あたり年間配当

    return code_to_name, code_to_ind, code_to_divps

# ============================================================
# #(6) 株価時系列（wide）：必要銘柄だけを「CSVをファイルごとに1回」読んで集約
#   - Data/kabuka/YYYY.csv は「先頭列が日付（idx）」想定
# ============================================================
def _read_kabuka_ts_wide_for_codes_detail(
    codes4: List[str],
    start_date: Optional[pd.Timestamp] = None,
    end_date: Optional[pd.Timestamp] = None,
) -> pd.DataFrame:
    if not codes4:
        return pd.DataFrame(columns=["date"])

    today = pd.Timestamp.today().normalize()
    if start_date is None:
        start_date = pd.Timestamp(year=today.year - 2, month=1, day=1)
    if end_date is None:
        end_date = today

    base_dir = os.path.join("Data", "kabuka")
    paths = sorted(glob.glob(os.path.join(base_dir, "[0-9][0-9][0-9][0-9].csv")))
    if not paths:
        return pd.DataFrame(columns=["date"] + codes4)

    chunks = []
    for p in paths:
        kdf = pd.read_csv(p, encoding="utf-8-sig")
        if len(kdf.columns) == 0:
            continue

        idx_col = "idx"
        first_col = kdf.columns[0]
        if first_col != idx_col:
            kdf = kdf.rename(columns={first_col: idx_col})

        present = [c for c in codes4 if c in kdf.columns]
        if not present:
            continue

        mask = kdf[idx_col].astype("string").str.match(r"^\d{4}-\d{2}-\d{2}$", na=False)
        tmp = kdf.loc[mask, [idx_col] + present].copy()

        tmp[idx_col] = pd.to_datetime(tmp[idx_col], errors="coerce")
        tmp = tmp.dropna(subset=[idx_col])
        if len(tmp) == 0:
            continue

        tmp = tmp[(tmp[idx_col] >= start_date) & (tmp[idx_col] <= end_date)]
        if len(tmp) == 0:
            continue

        for c in present:
            tmp[c] = pd.to_numeric(tmp[c], errors="coerce")

        tmp = tmp.rename(columns={idx_col: "date"})
        chunks.append(tmp)

    if not chunks:
        return pd.DataFrame(columns=["date"] + codes4)

    out = pd.concat(chunks, ignore_index=True)
    out = out.sort_values("date").drop_duplicates(subset=["date"], keep="last").reset_index(drop=True)

    for c in codes4:
        if c not in out.columns:
            out[c] = np.nan

    return out[["date"] + codes4]

# ============================================================
# #(7) 株価チャート（複数銘柄 / 指数化 / 凡例下）
# ============================================================
def _plot_price_multi_detail(
    ts_wide: pd.DataFrame,
    codes4: List[str],
    code_to_name: Dict[str, str],
    title: str,
):
    if ts_wide is None or len(ts_wide) == 0 or "date" not in ts_wide.columns:
        print("株価データが見つかりません（Data/kabuka のCSV/列名を確認）。")
        return

    plt.figure(figsize=(12, 5))
    plotted = 0

    for code in codes4:
        if code not in ts_wide.columns:
            continue

        s = ts_wide[["date", code]].copy().dropna(subset=["date"]).sort_values("date")
        s[code] = pd.to_numeric(s[code], errors="coerce")
        s = s.dropna(subset=[code])
        if len(s) == 0:
            continue

        base_val = float(s[code].iloc[0])
        if not (base_val > 0):
            continue

        y = s[code] / base_val
        label = f"{code} {code_to_name.get(code, '')}".strip()
        plt.plot(s["date"], y, label=label)
        plotted += 1

    if plotted == 0:
        print("株価チャート：対象期間にデータがありません（列名/期間/日付行を確認）。")
        return

    plt.title(title)
    plt.xlabel("日付")
    plt.ylabel("指数（期間先頭=1.0）")
    plt.legend(loc="upper center", bbox_to_anchor=(0.5, -0.18), ncol=3, fontsize=8, frameon=False)
    plt.subplots_adjust(bottom=0.28)
    plt.tight_layout()
    plt.show()

# ============================================================
# #(8) detail 本体（単一定義）
#   - codes: 推奨コードのlist（Top10をコピペ想定）
#   - owner: "T" or "Y"
# ============================================================
def detail(codes: List[str], owner: str):
    if not isinstance(codes, list) or len(codes) == 0:
        raise ValueError("detail は detail(['1301','1305'], 'T') のように list で銘柄コードを指定してください。")
    if owner is None or str(owner).strip() == "":
        raise ValueError("detail の owner は 'T' のように指定してください。")

    owner = str(owner).strip()

    owners_all = sorted(target_df["所有者"].astype("string").dropna().unique().tolist())
    if owner not in owners_all:
        raise ValueError(f"owner='{owner}' が target_df に見つかりません。存在する所有者: {owners_all}")

    codes4 = _norm_codes4([str(c) for c in codes])

    # マスタ（name/industry/divps）
    code_to_name, code_to_ind, code_to_divps = _build_master_maps_detail()

    # スコア用カテゴリ（ETF除外込み）
    idmap_df_for_cat = load_idmap(os.path.join("Data", "01_IDmap.csv"))
    all_categories_eval = load_all_categories_eval(idmap_df_for_cat, excluded=EXCLUDED_INDUSTRIES)

    # 最新株価（購入株数算定用）
    kabuka_path = find_latest_kabuka_file()
    kabuka_df = read_kabuka_wide(kabuka_path)

    # グラフ用：時系列（wide）
    ts_wide = _read_kabuka_ts_wide_for_codes_detail(codes4)

    # --------------------------------------------
    # 現状（owner）のDF：補*は apply_investment_supplement に一本化
    # --------------------------------------------
    df_owner = target_df.copy()
    df_owner["code"] = normalize_code_str_series(df_owner["code"]).map(_norm_code4)
    df_owner = df_owner[df_owner["所有者"].astype("string") == owner].copy()

    df_owner["株数"] = pd.to_numeric(df_owner["株数"], errors="coerce")
    df_owner["取得単価"] = pd.to_numeric(df_owner["取得単価"], errors="coerce")
    df_owner["年間配当"] = pd.to_numeric(df_owner["年間配当"], errors="coerce")

    df_owner["投資金額"] = df_owner["株数"] * df_owner["取得単価"]
    df_owner["配当金額"] = df_owner["株数"] * df_owner["年間配当"]

    # 業種ダミー除外（円グラフ/スコアの分母を汚さない）
    valid_ind = (
        df_owner["industry_33"].notna()
        & ~df_owner["industry_33"].astype("string").str.strip().isin(["", "-", "0"])
        & ~df_owner["industry_33"].astype("string").isin(EXCLUDED_INDUSTRIES)
    )
    df_owner = df_owner[valid_ind].copy()

    if len(df_owner) == 0:
        raise ValueError("ownerの保有が空になりました（industry_33欠損/除外により）。target_dfを確認してください。")

    # 補*列付与（Tのみ・益のみ控除を内部で実施）
    df_owner = apply_investment_supplement(
        df_owner,
        owner_col="所有者",
        code_col="code",
        name_col="name",
        invest_col="投資金額",
        sup_col="投資金額_補",
        realized_col="総合実現損益",
        summary_path=os.path.join("Data", "jyouto", "summary.csv"),
    )

    # owner != T の場合でも列互換を担保（共通セルの実装次第の保険）
    if owner != "T":
        df_owner["投資金額_補"] = df_owner["投資金額"]

    base = calc_portfolio_score(
        df_owner,
        all_categories_eval=all_categories_eval,
        industry_col="industry_33",
        code_col="code",
        invest_col_raw="投資金額",
        invest_col_sup="投資金額_補",
        dividend_amount_col="配当金額",
        use_sup_invest=True,
        yield_y0=4.0,
        yield_y1=10.0,
    )

    print("\n" + "=" * 100)
    print(f"detail：指定銘柄まとめ（owner={owner}）")
    print(f"株価参照: {os.path.basename(kabuka_path)}")
    print(
        f"現状スコア: {base['score_total']:.2f} / 100"
        f"（配当 {base['score_yield']:.2f}/40, 業種 {base['score_industry']:.2f}/40, 銘柄 {base['score_stock']:.2f}/20）"
    )
    print("=" * 100)

    # --------------------------------------------
    # 銘柄ごとの表（追加買いシミュレーション：owner分のポートフォリオに追加）
    # --------------------------------------------
    rows = []
    code_to_name_for_plot: Dict[str, str] = {}

    for code in codes4:
        name = code_to_name.get(code, "")
        ind = str(code_to_ind.get(code, "")).strip()
        divps = float(code_to_divps.get(code, 0.0))  # 1株配当（年）

        code_to_name_for_plot[code] = name

        # 業種が無効/除外ならスキップ
        if ind in ["", "-", "0"] or ind in EXCLUDED_INDUSTRIES:
            continue

        price = float(latest_price_from_kabuka(kabuka_df, code))
        buy_shares = int(calc_buy_shares(price))

        if not (price > 0) or buy_shares <= 0:
            continue

        buy_invest = float(buy_shares * price)
        annual_div_total = float(buy_shares * divps)

        y = (annual_div_total / buy_invest) * 100.0 if buy_invest > 0 else np.nan
        y_sup = y  # 追加分は実現益0なので、通常と同じ（補*はポートフォリオ分母側で効く）

        add_row = {
            "code": code,
            "name": name,
            "industry_33": ind,
            "株数": float(buy_shares),
            "取得単価": float(price),
            "年間配当": float(divps),
            "投資金額": float(buy_invest),
            "配当金額": float(annual_div_total),
            "投資金額_補": float(buy_invest),  # 追加分は実現益0
            "所有者": "ADD",
        }

        df_new = pd.concat([df_owner, pd.DataFrame([add_row])], ignore_index=True)

        after = calc_portfolio_score(
            df_new,
            all_categories_eval=all_categories_eval,
            industry_col="industry_33",
            code_col="code",
            invest_col_raw="投資金額",
            invest_col_sup="投資金額_補",
            dividend_amount_col="配当金額",
            use_sup_invest=True,
            yield_y0=4.0,
            yield_y1=10.0,
        )

        rows.append({
            "code": code,
            "name": name,
            "業種": ind,
            "年間配当(1株)": float(divps),
            "追加株数": int(buy_shares),
            "取得単価": float(price),
            "投資金額": float(buy_invest),
            "年間配当(合計)": float(annual_div_total),
            "配当利回り": float(y),
            "配当利回り(補*)": float(y_sup),
            "Δ総スコア": after["score_total"] - base["score_total"],
            "Δ配当": after["score_yield"] - base["score_yield"],
            "Δ業種": after["score_industry"] - base["score_industry"],
            "Δ銘柄": after["score_stock"] - base["score_stock"],
        })

    out = pd.DataFrame(rows)

    show_cols = [
        "code", "name", "業種",
        "年間配当(1株)",
        "追加株数", "取得単価", "投資金額",
        "年間配当(合計)", "配当利回り", "配当利回り(補*)",
        "Δ総スコア", "Δ配当", "Δ業種", "Δ銘柄",
    ]
    out = out[[c for c in show_cols if c in out.columns]].copy()

    fmt = {
        "年間配当(1株)": "{:.2f}",
        "追加株数": "{:,.0f}",
        "取得単価": "{:.2f}",
        "投資金額": "{:,.0f}",
        "年間配当(合計)": "{:,.0f}",
        "配当利回り": "{:.2f}",
        "配当利回り(補*)": "{:.2f}",
        "Δ総スコア": "{:+.2f}",
        "Δ配当": "{:+.2f}",
        "Δ業種": "{:+.2f}",
        "Δ銘柄": "{:+.2f}",
    }
    if len(out) == 0:
        print("指定銘柄がすべてスキップされました（業種除外/株価欠損/購入株数0 など）。")
    else:
        try:
            display(out.style.format(fmt))
        except Exception:
            display(out)

    # --------------------------------------------
    # グラフ：指数化＆凡例下
    # --------------------------------------------
    _plot_price_multi_detail(
        ts_wide=ts_wide,
        codes4=codes4,
        code_to_name=code_to_name_for_plot,
        title=f"株価（指数：期間先頭=1.0） / owner={owner} / codes={len(codes4)}",
    )
```

```python
# ============================================================
# #(4-1) 銘柄詳細：detail(["1301","1305"], "T") のように複数コードを一括表示（整理版・単一定義）
#   前提（上から順番に実行する想定）：
#     - target_df が存在（2つ目セルで作成済み）
#     - 3つ目セルで定義済みの関数が存在：
#         _build_haitou_pick_df
#         _find_latest_kabuka_file, _read_kabuka_wide, _latest_price_from_kabuka, _calc_buy_shares
#         _calc_portfolio_score
#     - EXCLUDED_INDUSTRIES, normalize_code_str_series, _to_num_scalar が（2/3つ目により）存在している想定
#
#   補*：
#     - owner == "T" のときのみ
#     - 実現損益 > 0 のときだけ投資金額を減算（損失は無視）
# ============================================================

import os
import re
import glob
from typing import List, Dict, Tuple, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


# ============================================================
# #(4-2) 4つ目セル内専用：コード正規化（lookup用）
#   - 全角→半角
#   - strip
#   - 数字のみで長さ<=4 は左ゼロ埋めで4桁化（kabuka列/IDmap/haitouに合わせやすく）
#   - 5桁以上（ETF等）はそのまま
# ============================================================
def _norm_code4(x: str) -> str:
    s = str(x).strip().translate(str.maketrans("０１２３４５６７８９", "0123456789"))
    if re.fullmatch(r"\d+", s) and len(s) <= 4:
        return s.zfill(4)
    return s

def _norm_codes4(codes: List[str]) -> List[str]:
    return [_norm_code4(c) for c in codes]


# ============================================================
# #(4-3) 依存チェック（足りないときはここで分かりやすく止める）
# ============================================================
_required_globals = [
    "target_df",
    "_build_haitou_pick_df",
    "_find_latest_kabuka_file", "_read_kabuka_wide", "_latest_price_from_kabuka", "_calc_buy_shares",
    "_calc_portfolio_score",
]
_missing = [name for name in _required_globals if name not in globals()]
if _missing:
    raise NameError(
        "4つ目セルの前提が不足しています。上から順に実行できているか確認してください。不足: "
        + ", ".join(_missing)
    )

# EXCLUDED_INDUSTRIES が無い場合だけ保険
if "EXCLUDED_INDUSTRIES" not in globals():
    EXCLUDED_INDUSTRIES = {"日経ETF", "米国ETF"}  # fallback


# ============================================================
# #(4-4) マスタ作成：code -> (name, industry_33, 年間配当)
#   - IDmap + haitou_pick を結合して辞書化（detailで繰り返し読まない）
# ============================================================
def _build_candidate_master_maps_detail() -> Tuple[Dict[str, str], Dict[str, str], Dict[str, float]]:
    idmap_df = pd.read_csv("Data/01_IDmap.csv", dtype={"code": "string"}, encoding="utf-8-sig")
    # BOM対策
    if "code" not in idmap_df.columns:
        for c in idmap_df.columns:
            if str(c).replace("\ufeff", "") == "code":
                idmap_df = idmap_df.rename(columns={c: "code"})
                break
    if "code" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に code 列が見つかりません。")
    if "industry_33" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に industry_33 列が見つかりません。")
    if "name" not in idmap_df.columns:
        idmap_df["name"] = ""

    idmap_df["code"] = idmap_df["code"].astype("string").map(lambda v: _norm_code4(v))

    haitou_pick_df = _build_haitou_pick_df()
    haitou_pick_df["code"] = haitou_pick_df["code"].astype("string").map(lambda v: _norm_code4(v))
    haitou_pick_df["年間配当"] = pd.to_numeric(haitou_pick_df["年間配当"], errors="coerce").fillna(0.0)

    m = pd.merge(
        idmap_df[["code", "name", "industry_33"]],
        haitou_pick_df[["code", "年間配当"]],
        on="code",
        how="left",
    )
    m["年間配当"] = pd.to_numeric(m["年間配当"], errors="coerce").fillna(0.0)

    code_to_name = dict(zip(m["code"].astype(str), m["name"].fillna("").astype(str)))
    code_to_ind = dict(zip(m["code"].astype(str), m["industry_33"].fillna("").astype(str)))
    code_to_div = dict(zip(m["code"].astype(str), m["年間配当"].astype(float)))

    return code_to_name, code_to_ind, code_to_div


# ============================================================
# #(4-5) summary.csv（2行目）読み込み：実現損益map（1回だけ）
#   - key: 4桁コード or 銘柄名
# ============================================================
def _load_realized_pl_maps_once_detail() -> Tuple[Dict[str, float], Dict[str, float]]:
    summary_path = os.path.join("Data", "jyouto", "summary.csv")
    if not os.path.exists(summary_path):
        return {}, {}

    sdf = pd.read_csv(summary_path, dtype="string", encoding="utf-8-sig")
    if len(sdf) < 2:
        return {}, {}

    row2 = sdf.iloc[1]
    pl_by_code: Dict[str, float] = {}
    pl_by_name: Dict[str, float] = {}

    # _to_num_scalar が無い場合の保険
    def _to_num_scalar_local(x) -> float:
        if pd.isna(x):
            return float("nan")
        s = str(x).replace(",", "").strip()
        m = re.search(r"-?\d+(?:\.\d+)?", s)
        if not m:
            return float("nan")
        try:
            return float(m.group(0))
        except Exception:
            return float("nan")

    to_num = _to_num_scalar if ("_to_num_scalar" in globals()) else _to_num_scalar_local

    for k, v in row2.items():
        if k is None:
            continue
        key_raw = str(k).strip()
        if key_raw == "":
            continue

        val = to_num(v)
        if pd.isna(val):
            continue

        key_norm = _norm_code4(key_raw)
        if re.fullmatch(r"\d{4}", key_norm):
            pl_by_code[key_norm] = float(val)
        else:
            pl_by_name[key_raw] = float(val)

    return pl_by_code, pl_by_name

def _get_realized_pl(code4: str, name: str, pl_by_code: Dict[str, float], pl_by_name: Dict[str, float]) -> float:
    if code4 in pl_by_code:
        return float(pl_by_code[code4])
    if name in pl_by_name:
        return float(pl_by_name[name])
    return 0.0


# ============================================================
# #(4-6) スコア用カテゴリ（industry_33 全種類：ETF2業種除外）
# ============================================================
def _load_all_categories_eval_detail() -> List[str]:
    idmap_df = pd.read_csv("Data/01_IDmap.csv", dtype={"code": "string"}, encoding="utf-8-sig")
    if "industry_33" not in idmap_df.columns:
        raise KeyError("01_IDmap.csv に industry_33 列が見つかりません。")

    cats = (
        idmap_df["industry_33"]
        .astype("string")
        .map(lambda v: v.strip() if isinstance(v, str) else v)
        .dropna()
    )
    cats = sorted({v for v in cats.unique().tolist() if v not in ["", "-", "0"]})
    cats_eval = [c for c in cats if c not in set(EXCLUDED_INDUSTRIES)]
    if len(cats_eval) == 0:
        raise ValueError("評価対象の industry_33 が 0 件になりました。除外設定やIDmapを確認してください。")
    return cats_eval


# ============================================================
# #(4-7) 株価時系列（wide）：必要銘柄だけを「CSVをファイルごとに1回」読んで集約
# ============================================================
def _read_kabuka_ts_wide_for_codes_detail(
    codes4: List[str],
    start_date: Optional[pd.Timestamp] = None,
    end_date: Optional[pd.Timestamp] = None,
) -> pd.DataFrame:
    if not codes4:
        return pd.DataFrame(columns=["date"])

    today = pd.Timestamp.today().normalize()
    if start_date is None:
        start_date = pd.Timestamp(year=today.year - 2, month=1, day=1)
    if end_date is None:
        end_date = today

    base_dir = os.path.join("Data", "kabuka")
    paths = sorted(glob.glob(os.path.join(base_dir, "[0-9][0-9][0-9][0-9].csv")))
    if not paths:
        return pd.DataFrame(columns=["date"] + codes4)

    chunks = []
    for p in paths:
        kdf = pd.read_csv(p, encoding="utf-8-sig")
        first_col = kdf.columns[0]
        idx_col = "idx"
        if first_col != idx_col:
            kdf = kdf.rename(columns={first_col: idx_col})

        present = [c for c in codes4 if c in kdf.columns]
        if not present:
            continue

        mask = kdf[idx_col].astype("string").str.match(r"^\d{4}-\d{2}-\d{2}$", na=False)
        tmp = kdf.loc[mask, [idx_col] + present].copy()
        tmp[idx_col] = pd.to_datetime(tmp[idx_col], errors="coerce")
        tmp = tmp.dropna(subset=[idx_col])
        if len(tmp) == 0:
            continue

        tmp = tmp[(tmp[idx_col] >= start_date) & (tmp[idx_col] <= end_date)]
        if len(tmp) == 0:
            continue

        for c in present:
            tmp[c] = pd.to_numeric(tmp[c], errors="coerce")

        tmp = tmp.rename(columns={idx_col: "date"})
        chunks.append(tmp)

    if not chunks:
        return pd.DataFrame(columns=["date"] + codes4)

    out = pd.concat(chunks, ignore_index=True)
    out = out.sort_values("date").drop_duplicates(subset=["date"], keep="last").reset_index(drop=True)

    for c in codes4:
        if c not in out.columns:
            out[c] = np.nan

    return out[["date"] + codes4]


def _plot_price_multi_detail(
    ts_wide: pd.DataFrame,
    codes4: List[str],
    code_to_name: Dict[str, str],
    title: str,
):
    if ts_wide is None or len(ts_wide) == 0 or "date" not in ts_wide.columns:
        print("株価データが見つかりません（Data/kabuka のCSV/列名を確認）。")
        return

    today = pd.Timestamp.today().normalize()
    base_date = pd.Timestamp(year=today.year - 2, month=1, day=1)

    plt.figure(figsize=(12, 5))
    plotted = 0

    for code in codes4:
        if code not in ts_wide.columns:
            continue
        s = ts_wide[["date", code]].copy().dropna(subset=["date"]).sort_values("date")
        s[code] = pd.to_numeric(s[code], errors="coerce")
        s = s.dropna(subset=[code])
        if len(s) == 0:
            continue

        s_after = s[s["date"] >= base_date]
        if len(s_after) == 0:
            continue
        base_val = float(s_after[code].iloc[0])
        if not (base_val > 0):
            continue

        y = s_after[code] / base_val
        label = f"{code} {code_to_name.get(code, '')}".strip()
        plt.plot(s_after["date"], y, label=label)
        plotted += 1

    if plotted == 0:
        print("株価チャート：対象期間にデータがありません（列名/期間/日付行を確認）。")
        return

    plt.title(title)
    plt.xlabel("日付")
    plt.ylabel("指数（基準=今年-2年の1/1を1.0）")
    plt.legend(loc="upper center", bbox_to_anchor=(0.5, -0.18), ncol=3, fontsize=8, frameon=False)
    plt.subplots_adjust(bottom=0.28)
    plt.tight_layout()
    plt.show()


# ============================================================
# #(4-8) detail 本体（単一定義）
#   - codes: 推奨コードのlist（3つ目で出した Top10 をコピペ想定）
#   - owner: "T" or "Y"
# ============================================================
def detail(codes: List[str], owner: str):
    if not isinstance(codes, list) or len(codes) == 0:
        raise ValueError("detail は detail(['1301','1305'], 'T') のように list で銘柄コードを指定してください。")
    if owner is None or str(owner).strip() == "":
        raise ValueError("detail の owner は 'T' のように指定してください。")

    owner = str(owner).strip()

    owners_all = sorted(target_df["所有者"].astype("string").dropna().unique().tolist())
    if owner not in owners_all:
        raise ValueError(f"owner='{owner}' が target_df に見つかりません。存在する所有者: {owners_all}")

    codes4 = _norm_codes4([str(c) for c in codes])

    # マスタ/実現損益/カテゴリ
    code_to_name, code_to_ind, code_to_divps = _build_candidate_master_maps_detail()
    pl_by_code, pl_by_name = _load_realized_pl_maps_once_detail()
    all_categories_eval = _load_all_categories_eval_detail()

    # 最新株価参照（購入株数算定用）
    kabuka_path = _find_latest_kabuka_file()
    kabuka_df = _read_kabuka_wide(kabuka_path)

    # グラフ用：時系列（wide）
    ts_wide = _read_kabuka_ts_wide_for_codes_detail(codes4)

    # --------------------------------------------
    # 現状（owner）のDFを組む：補*は owner=="T" のときのみ
    # --------------------------------------------
    df_owner = target_df.copy()
    df_owner["code"] = df_owner["code"].astype("string").map(lambda v: _norm_code4(v))
    df_owner = df_owner[df_owner["所有者"].astype("string") == owner].copy()

    df_owner["株数"] = pd.to_numeric(df_owner["株数"], errors="coerce")
    df_owner["取得単価"] = pd.to_numeric(df_owner["取得単価"], errors="coerce")
    df_owner["年間配当"] = pd.to_numeric(df_owner["年間配当"], errors="coerce")

    df_owner["投資金額"] = df_owner["株数"] * df_owner["取得単価"]
    df_owner["配当金額"] = df_owner["株数"] * df_owner["年間配当"]

    # 業種ダミー除外
    valid_ind = (
        df_owner["industry_33"].notna()
        & ~df_owner["industry_33"].astype("string").str.strip().isin(["", "-", "0"])
    )
    df_owner = df_owner[valid_ind].copy()

    # 補*列：投資金額_補（owner=="T" かつ 実現益>0 のときだけ控除）
    if owner == "T":
        df_owner["name"] = df_owner.get("name", "")
        df_owner["_pl"] = df_owner.apply(
            lambda r: _get_realized_pl(str(r["code"]), str(r.get("name", "")), pl_by_code, pl_by_name),
            axis=1
        )
        df_owner["_pl_pos"] = np.where(df_owner["_pl"] > 0, df_owner["_pl"], 0.0)
        df_owner["投資金額_補"] = df_owner["投資金額"] - df_owner["_pl_pos"]
    else:
        df_owner["投資金額_補"] = df_owner["投資金額"]  # 互換のため列は持たせる

    # 現状スコア
    base = _calc_portfolio_score(df_owner, all_categories_eval)

    print("\n" + "=" * 100)
    print(f"detail：指定銘柄まとめ（owner={owner}）")
    print(f"株価参照: {os.path.basename(kabuka_path)}")
    print(
        f"現状スコア: {base['スコア_合計(100)']:.2f} / 100"
        f"（配当 {base['スコア_配当利回り(40)']:.2f}/40, 業種 {base['スコア_業種(40)']:.2f}/40, 銘柄 {base['スコア_銘柄(20)']:.2f}/20）"
    )
    print("=" * 100)

    # --------------------------------------------
    # 銘柄ごとの表（1枚）
    # --------------------------------------------
    rows = []
    code_to_name_for_plot: Dict[str, str] = {}

    for code in codes4:
        name = code_to_name.get(code, "")
        ind = code_to_ind.get(code, "")
        divps = float(code_to_divps.get(code, 0.0))  # 1株配当（年）

        code_to_name_for_plot[code] = name

        price = float(_latest_price_from_kabuka(kabuka_df, code))
        buy_shares = int(_calc_buy_shares(price))
        buy_invest = float(buy_shares * price) if (buy_shares > 0 and price > 0) else float("nan")

        realized_pl = float(_get_realized_pl(code, name, pl_by_code, pl_by_name))

        # 年間配当（合計）＝株数×1株配当
        annual_div_total = float(buy_shares * divps) if buy_shares > 0 else 0.0

        # 配当利回り（通常）
        y = np.nan
        if pd.notna(buy_invest) and buy_invest > 0:
            y = (annual_div_total / buy_invest) * 100.0

        # 配当利回り（補*）：owner=="T" のときのみ、実現益（>0）を控除して分母を下げる
        invest_sup = buy_invest
        if owner == "T" and realized_pl > 0 and pd.notna(buy_invest):
            invest_sup = buy_invest - realized_pl

        y_sup = np.nan
        if pd.notna(invest_sup) and invest_sup > 0:
            y_sup = (annual_div_total / invest_sup) * 100.0

        # 追加後スコア（ownerのみ）
        add_row = {
            "code": code,
            "industry_33": ind,
            "株数": float(buy_shares),
            "取得単価": float(price) if (price > 0) else np.nan,
            "年間配当": float(divps),
            "投資金額": float(buy_shares * price) if (buy_shares > 0 and price > 0) else 0.0,
            "配当金額": float(buy_shares * divps) if (buy_shares > 0) else 0.0,
            "投資金額_補": float(buy_shares * price) if (buy_shares > 0 and price > 0) else 0.0,  # 追加分は実現益0
            "所有者": "ADD",
            "name": name,
        }
        df_new = pd.concat([df_owner, pd.DataFrame([add_row])], ignore_index=True)
        after = _calc_portfolio_score(df_new, all_categories_eval)

        rows.append({
            "code": code,
            "name": name,
            "業種": ind,
            "年間配当(1株)": float(divps),
            "実現損益": float(realized_pl),
            "追加株数": int(buy_shares),
            "取得単価": float(price) if (price > 0) else np.nan,
            "投資金額": float(buy_invest),
            "年間配当(合計)": float(annual_div_total),
            "配当利回り": float(y) if pd.notna(y) else np.nan,
            "配当利回り(補*)": float(y_sup) if pd.notna(y_sup) else np.nan,
            "Δ総スコア": after["スコア_合計(100)"] - base["スコア_合計(100)"],
            "Δ配当": after["スコア_配当利回り(40)"] - base["スコア_配当利回り(40)"],
            "Δ業種": after["スコア_業種(40)"] - base["スコア_業種(40)"],
            "Δ銘柄": after["スコア_銘柄(20)"] - base["スコア_銘柄(20)"],
        })

    out = pd.DataFrame(rows)

    show_cols = [
        "code", "name", "業種",
        "年間配当(1株)", "実現損益",
        "追加株数", "取得単価", "投資金額",
        "年間配当(合計)", "配当利回り", "配当利回り(補*)",
        "Δ総スコア", "Δ配当", "Δ業種", "Δ銘柄",
    ]
    show_cols = [c for c in show_cols if c in out.columns]
    out = out[show_cols].copy()

    fmt = {
        "年間配当(1株)": "{:.2f}",
        "実現損益": "{:,.0f}",
        "追加株数": "{:,.0f}",
        "取得単価": "{:.2f}",
        "投資金額": "{:,.0f}",
        "年間配当(合計)": "{:,.0f}",
        "配当利回り": "{:.2f}",
        "配当利回り(補*)": "{:.2f}",
        "Δ総スコア": "{:+.2f}",
        "Δ配当": "{:+.2f}",
        "Δ業種": "{:+.2f}",
        "Δ銘柄": "{:+.2f}",
    }
    try:
        display(out.style.format(fmt))
    except Exception:
        display(out)

    # --------------------------------------------
    # グラフ（1枚）：指数化＆凡例下
    # --------------------------------------------
    _plot_price_multi_detail(
        ts_wide=ts_wide,
        codes4=codes4,
        code_to_name=code_to_name_for_plot,
        title=f"株価（指数：{pd.Timestamp.today().year-2}-01-01=1.0） / owner={owner} / codes={len(codes4)}",
    )
```
